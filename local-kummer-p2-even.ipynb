# The local Kummer image at $p = 2$ for $y^2 = x^3 + b$ with $v_2(b) = 1$

The four curves $b \in \{2,\ -2,\ 6,\ -6\}$, treated together because at $2$ they share one and the
same descent field — and, as it turns out, one and the same image.

**Why $v_2(b) = 1$ changes everything.** In the companion notebooks $b$ was a *unit*, every $2$-adic
unit is a cube, and $x^3+b$ therefore acquired a root in $\mathbb{Q}_2$: the algebra split as
$\mathbb{Q}_2 \times \mathbb{Q}_4$, $E[2] = \operatorname{Ind}_{\mathbb{Q}_4}^{\mathbb{Q}_2}\mu_2$, and Shapiro
turned the target into $\mathbb{Q}_4^\times/(\mathbb{Q}_4^\times)^2$, of order $16$. Now
$v_2(-b) = 1$ is **not divisible by $3$**, so $-b$ is not a cube, $x^3+b$ is *Eisenstein* at $2$, and
$$L \;=\; \mathbb{Q}_2[x]/(x^3+b) \;=\; \mathbb{Q}_2(\pi), \qquad \pi = \sqrt[3]{-b},$$
is a **totally ramified cubic field** ($e = 3$, $f = 1$), with $\pi$ a uniformiser. The Galois group
of the splitting field $L(\omega)$ is the full $S_3$, so $E[2]$ has no rational point at all:
$$E[2](\mathbb{Q}_2) = 0, \qquad \#E(\mathbb{Q}_2)/2E(\mathbb{Q}_2) = 2^{r_2+1} = 2 .$$
The target is no longer one field's square classes but the norm-one part of $L^\times/(L^\times)^2$:
$$H^1(\mathbb{Q}_2, E[2]) \;\cong\; \ker\big(N_{L/\mathbb{Q}_2} : L^\times/(L^\times)^2 \to
\mathbb{Q}_2^\times/(\mathbb{Q}_2^\times)^2\big), \qquad \text{of order } 4,$$
and $\operatorname{Im}(\delta)$ is a maximal isotropic subgroup of order $2$ in it. Everything is
smaller than before, and correspondingly there is exactly **one** non-trivial class to compute per
curve.

**One field for all four.** $\mathbb{Q}_2(\sqrt[3]{-b}) = \mathbb{Q}_2(\sqrt[3]{2})$ whenever
$v_2(b) = 1$, because $-b/2$ is a unit, hence a cube. So we fix once and for all
$$L = \mathbb{Q}_2(\pi), \quad \pi = \sqrt[3]{2}, \qquad
\delta_b(x,y) = x - \lambda_b\,\pi, \quad \lambda_b = \sqrt[3]{-b/2} \in \mathbb{Z}_2^\times,$$
which is the canonical, $\mathbb{Q}_2$-rational identification of the four $E_b[2]$ used in the
companion notebooks. Here $\lambda_2 = -1$, $\lambda_{-2} = 1$, $\lambda_6 = \sqrt[3]{-3}$ and
$\lambda_{-6} = \sqrt[3]{3}$ — the same two irrational cube roots as in the $b = \pm3$ notebooks, and
replaceable by $5$ and $3$ modulo squares by the same lemma (§4).

**The answer.** With $V = \{\mathcal{O},\ P\}$ and

| $b$ | $P$ | $\delta_b(P)$ |
|---|---|---|
| $2$ | $(-1,\ 1)$ | $\pi-1$ |
| $-2$ | $(3,\ 5)$ | $3-\pi$ |
| $6$ | $(3,\ \sqrt{33})$ | $3-5\pi$ |
| $-6$ | $(-1,\ \sqrt{-7})$ | $-1-3\pi$ |

**all four give the same class**, and
$$\operatorname{Im}(\delta_b) \;=\; \big\{1,\ [-5(1+\pi)]\big\} \;\subset\;
\ker N = \big\{1,\ [-(1+\pi)(1+\pi^2)],\ [-5(1+\pi)],\ [5(1+\pi^2)]\big\} .$$
That is not forced: the four curves are pairwise non-isomorphic over $\mathbb{Q}_2$, all three
order-$2$ subgroups of $\ker N$ are maximal isotropic, and the four have different ranks over
$\mathbb{Q}$ ($1,1,0,0$). Contrast the unit family, where the four curves realised *two* Lagrangians,
exchanged by Frobenius (§8).

Companions: `local-kummer-p2.ipynb`, `-minus`, `-plus3`, `-minus3` treat $b = 1, -1, 3, -3$.

## 1. The four curves at $2$

```python
R.<X> = QQ[]

def label(E):
    try:    return E.cremona_label()
    except Exception: return "(not in db)"

print(" b   label       disc          conductor    Kodaira c_2   E(Q)")
for b in [2, -2, 6, -6]:
    E  = EllipticCurve([0,0,0,0,b])
    ld = E.local_data(2)
    mw = "rank " + str(E.rank()) + ", tors " + str(E.torsion_subgroup().invariants())
    print(str(b).rjust(2), label(E).rjust(12), str(factor(E.discriminant())).rjust(14),
          str(factor(E.conductor())).rjust(11), "  ", str(ld.kodaira_symbol()).ljust(4),
          str(ld.tamagawa_number()).rjust(2), "  " + mw,
          "  gens", [g.xy() for g in E.gens()])
```

```python
Q2 = Qp(2, 60)
print("v_2(-b) = 1 is not divisible by 3, so -b is not a cube in Q_2 and x^3+b is Eisenstein:")
for b in [2, -2, 6, -6]:
    L0 = NumberField(X^3 + b, 'p')
    print("  b =", str(b).rjust(2),
          "  x^3+b irreducible over Q_2 ?", (X^3+b).change_ring(Q2).is_irreducible(),
          "   primes above 2 in Q(cbrt(-b)), as (e,f):",
          [(p.ramification_index(), p.residue_class_degree()) for p in L0.primes_above(2)],
          "   Galois closure degree:", L0.galois_closure('a').degree())
print()
print("=> E[2](Q_2) = 0, r_2 = 0, and #E(Q_2)/2E(Q_2) = #E(Q_2)[2] * |2|_2^(-1) = 1 * 2 = 2")
```

All four have additive reduction of type $\mathrm{II}$ with $c_2 = 1$, so $E(\mathbb{Q}_2) = E_0$,
$E_0/E_1 \cong \tilde E_{\mathrm{ns}}(\mathbb{F}_2) = (\mathbb{F}_2,+)$ and $E_1 \cong \mathbb{Z}_2$;
hence $E(\mathbb{Q}_2)$ is pro-$2$, and it is torsion free because any torsion would be $2$-power
while $E[2](\mathbb{Q}_2) = 0$. So
$$E(\mathbb{Q}_2) \;\cong\; \mathbb{Z}_2, \qquad E_1 \text{ has index } 2, \qquad
2E(\mathbb{Q}_2) = E_1 ,$$
the last step because $\mathbb{Z}_2$ has a unique subgroup of index $2$. The dichotomy is therefore as
clean as it can be:
$$\delta_b(x,y) = 1 \iff (x,y) \in E_1 \iff v_2(x) < 0 ,$$
and every point with **integral** $x$ represents the non-trivial coset. §6 confirms this by scan.

**One field for all four $b$.** $\mathbb{Q}_2(\sqrt[3]{-b})$ depends only on $v_2(b)$ here, since
$-b/2$ is a unit and units are cubes:

```python
Z2 = Zp(2, 60)
print("cbrt(-b/2) for the four b, all 2-adic units:")
for b in [2, -2, 6, -6]:
    lam = Z2(QQ(-b)/2).nth_root(3)
    print("   b =", str(b).rjust(2), "   -b/2 =", str(QQ(-b)/2).rjust(2),
          "   lambda_b = cbrt(-b/2) =", str(lam.lift() % 128).rjust(3), "mod 128",
          "   lambda^3 + b/2 =", lam^3 - QQ(-b)/2)
print()
print("so Q_2(cbrt(-b)) = Q_2(cbrt 2) for all four:  cbrt(-b) = lambda_b * cbrt(2)")
```

## 2. The ramified cubic field $L = \mathbb{Q}_2(\sqrt[3]{2})$ and its square classes

$L$ has $e = 3$, $f = 1$, uniformiser $\pi$ with $\pi^3 = 2$, so $v_L(2) = 3$. By the same structure
theory as before, for $K/\mathbb{Q}_2$ of degree $d$,
$$\#K^\times/(K^\times)^2 = 2^{\,d+2},$$
so here $\#L^\times/(L^\times)^2 = 2^5 = 32$: one class of odd valuation, $[-1]$, and *three* unit
classes (against one, one and two for $\mathbb{Q}_4$).

The mod-$8$ criterion becomes a mod-$\mathfrak{p}^7$ criterion: $U^{(m)} \subset (L^\times)^2$ for
$m > 2e/(p-1) = 6$. And $\mathfrak{p}^7 = \pi^7\mathcal{O} = 4\pi\mathcal{O} =
8\mathbb{Z}_2 \oplus 4\pi\mathbb{Z}_2 \oplus 4\pi^2\mathbb{Z}_2$, so
$$\mathcal{O}/\mathfrak{p}^7 \;\cong\; \mathbb{Z}/8 \times \mathbb{Z}/4 \times \mathbb{Z}/4,
\qquad a + b\pi + d\pi^2 \ \longmapsto\ (a \bmod 8,\ b \bmod 4,\ d \bmod 4),$$
a ring of $128$ elements in which squareness of a unit can be decided by hand:

```python
def mulp7(u, v):
    "product in O/p^7 = (Z/8 x Z/4 x Z/4), elements (a,b,d) = a + b*pi + d*pi^2, pi^3 = 2"
    (a, b, d), (A, B, D) = u, v
    return ((a*A + 2*(b*D + B*d)) % 8, (a*B + A*b + 2*d*D) % 4, (a*D + A*d + b*B) % 4)

els   = [(a, b, d) for a in range(8) for b in range(4) for d in range(4)]
units = [u for u in els if u[0] % 2 == 1]
sqs   = sorted(set(mulp7(u, u) for u in units))

print("#O/p^7     =", len(els))
print("#(O/p^7)^* =", len(units))
print("squares    =", len(sqs), " namely", sqs)
print("index      =", len(units)//len(sqs), " = #O^*/(O^*)^2")
print("times the Z/2 from the valuation:  #L^*/(L^*)^2 =", 2*len(units)//len(sqs))
```

Symbolic bookkeeping happens in the exact cubic field $\mathbb{Q}(\sqrt[3]{2})$, in which $2$ is
totally ramified, so its completion at the unique prime above $2$ *is* $L$; squareness is decided in
Sage's Eisenstein extension `K.extension(t^3-2)`. A basis of $L^\times/(L^\times)^2$:
$$B = \big(\pi,\ -1,\ 1+\pi,\ 1+\pi^2,\ 5\big).$$
Note $5$ is a genuine basis element here — in $\mathbb{Q}_4$ it was a square. In fact
$\mathbb{Q}_2^\times/(\mathbb{Q}_2^\times)^2 \hookrightarrow L^\times/(L^\times)^2$ because
$[L:\mathbb{Q}_2] = 3$ is odd: if $a \in \mathbb{Q}_2^\times$ is a square in $L$ then so is
$N(a) = a^3$, hence so is $a$.

```python
K = Qp(2, 80); S.<t> = K[]
L.<pi>  = K.extension(t^3 - 2)          # the completion:  pi = cbrt(2)
L0.<p0> = NumberField(X^3 - 2)          # the exact home

def to2(z):
    "image of z in L under p0 |-> pi"
    return sum(L(z.list()[i]) * pi^i for i in range(3))

def is_sq(z):
    "is z a square in L ?"
    return to2(z).is_square()

def cls(q):
    "the square class of a rational q in Q_2^*, as (v_2(q) mod 2, unit part mod 8)"
    q = QQ(q); v = q.valuation(2); u = q / 2^v
    return (v % 2, ZZ(u.numerator() * ZZ(u.denominator()).inverse_mod(8)) % 8)

B      = [p0, L0(-1), 1 + p0, 1 + p0^2, L0(5)]
Bnames = ["pi", "-1", "1+pi", "1+pi^2", "5"]

def rep(m):
    z = L0(1)
    for i in range(5):
        if (m >> i) & 1:
            z *= B[i]
    return z

def coords(z):
    for m in range(32):
        if is_sq(z * rep(m)):
            return tuple((m >> i) & 1 for i in range(5))
    raise ValueError("z is not in the span of B")

print("L =", L, "   e =", L.e(), "  f =", L.f(), "   v_L(2) =", L(2).valuation())
print("basis:", Bnames)
print("independent (only the empty product is a square) ?",
      [m for m in range(32) if is_sq(rep(m))] == [0])
print()
print("Q_2^*/squares injects (odd degree):",
      [(a, is_sq(L0(a))) for a in [-1, 2, 5, 3, 7, -3, -7, 17]])
```

## 3. $H^1(\mathbb{Q}_2, E[2]) = \ker N$, of order $4$

From $E[2] \cong \{(a_1,a_2,a_3) \in \mu_2^3 : a_1a_2a_3 = 1\}$ with $\operatorname{Gal}$ permuting the
coordinates, there is an exact sequence of Galois modules
$$0 \to E[2] \to \operatorname{Res}_{L/\mathbb{Q}_2}\mu_2 \xrightarrow{\ N\ } \mu_2 \to 0 .$$
On $H^0$ the map $\mu_2 \to \mu_2$ is $a \mapsto a^3 = a$, surjective, so
$H^1(\mathbb{Q}_2,E[2]) \cong \ker\big(N : L^\times/(L^\times)^2 \to \mathbb{Q}_2^\times/(\mathbb{Q}_2^\times)^2\big)$
— using Shapiro for $H^1(\mathbb{Q}_2,\operatorname{Res}\mu_2) = L^\times/(L^\times)^2$. The order is
forced by the local Euler characteristic: $\#H^0 = \#H^2 = \#E[2](\mathbb{Q}_2) = 1$ ($E[2]$ is
self-dual) and $\#H^0\#H^2/\#H^1 = \|4\|_2 = 1/4$, so $\#H^1 = 4$. Equivalently $32/8 = 4$, since
$N$ is onto ($N(a) = a^3 \sim a$ for $a \in \mathbb{Q}_2^\times$).

```python
ker = [m for m in range(32) if Q2(QQ(rep(m).norm())).is_square()]
print("#ker N =", len(ker), "   (= #H^1(Q_2, E[2]))")
print()
print("  coords                representative      norm    norm mod 8")
for m in ker:
    z = rep(m)
    print("  " + str(tuple((m >> i) & 1 for i in range(5))) + "   " + str(z).rjust(18)
          + "   " + str(z.norm()).rjust(6) + "      " + str(ZZ(z.norm()) % 8))
print()
u1, u2, u3 = -(1+p0)*(1+p0^2), -5*(1+p0), 5*(1+p0^2)
print("readable form:  u1 = -(1+pi)(1+pi^2) =", u1, ",  u2 = -5(1+pi),  u3 = 5(1+pi^2)")
print("N(1+pi) =", (1+p0).norm(), "  N(1+pi^2) =", (1+p0^2).norm(),
      "   so N(u1),N(u2),N(u3) =", u1.norm(), u2.norm(), u3.norm())
print("u1*u2*u3 a square in L ?", is_sq(u1*u2*u3), "  => ker N = (Z/2)^2")
```

## 4. The descent map in a common frame

$\delta_b(x,y) = x - \sqrt[3]{-b} = x - \lambda_b\pi$ with $\lambda_b = \sqrt[3]{-b/2}$, and
$N_{L/\mathbb{Q}_2}\big(\delta_b(x,y)\big) = x^3+b = y^2$ is automatically a square, so the image does
land in $\ker N$. For $b = \pm2$ the scalar is rational ($\lambda_2 = -1$, $\lambda_{-2} = 1$); for
$b = \pm 6$ it is $\sqrt[3]{\mp3}$, the very cube roots met in the $b = \pm3$ notebooks, and the same
substitution lemma applies — in its ramified form.

> **Lemma.** Let $z, \tilde z \in L^\times$ with $v_L(\tilde z - z) \geq v_L(z) + 7$. Then
> $\tilde z/z \in U^{(7)} \subset (L^\times)^2$, so $[\tilde z] = [z]$.

For $z = x - \lambda\pi$ with $x \in \mathbb{Q}_2$ we have $v_L(z) = \min(3v_2(x), 1) \leq 1$, and
$v_L\big((\lambda-\tilde\lambda)\pi\big) = 3\,v_2(\lambda-\tilde\lambda) + 1$, so the hypothesis holds
as soon as $3v_2(\lambda-\tilde\lambda) + 1 \geq 8$, i.e. $v_2(\lambda-\tilde\lambda) \geq 3$. Since
$v_2(\sqrt[3]{-3} - 5) = 7$ and $v_2(\sqrt[3]{3} - 3) = 3$, we may take
$$\lambda_2 = -1, \qquad \lambda_{-2} = 1, \qquad \lambda_6 \rightsquigarrow 5, \qquad
\lambda_{-6} \rightsquigarrow 3 .$$
After the substitution the norm is no longer literally $y^2$, but it stays in the same square class —
which is all that matters.

```python
lam = {2: -1, -2: 1, 6: 5, -6: 3}
e2, ph = Z2(-3).nth_root(3), Z2(3).nth_root(3)
print("v_2(cbrt(-3) - 5) =", (e2 - 5).valuation(), "   v_2(cbrt(3) - 3) =", (ph - 3).valuation(),
      "   (both >= 3, the bound the lemma needs)")
print()
xP = {2: -1, -2: 3, 6: 3, -6: -1}
for b, lm in [(6, e2), (-6, ph)]:
    x0 = xP[b]
    z_true = L(x0) - to2(p0) * L(lm)                 # honest x - cbrt(-b)
    z_sub  = to2(L0(x0) - lam[b]*p0)                 # x - lambda~ * pi
    print("b =", str(b).rjust(2), ":  [x - cbrt(-b)] == [x -", lam[b], "* pi] ?",
          (z_true * z_sub).is_square())
```

## 5. The points and the image

$E(\mathbb{Q}_2)/2E(\mathbb{Q}_2)$ has order $2$, so $V = \{\mathcal{O}, P\}$ for any $P$ with
integral $x$ (§1). For $b = \pm2$ a rational generator is available; for $b = \pm6$ the curve has
rank $0$ and no torsion, so $E(\mathbb{Q}) = \{\mathcal{O}\}$ and $P$ must be found $2$-adically —
$x$ still rational, $y$ not.

```python
print(" b    P                        delta = x - lam*pi      N        N mod 8   coords")
img = {}
for b in [2, -2, 6, -6]:
    x0 = xP[b]; v = x0^3 + b
    z  = L0(x0) - lam[b]*p0
    img[b] = coords(z)
    nm = ZZ(z.norm())
    print(str(b).rjust(2), (" P = (" + str(x0) + ", sqrt(" + str(v) + "))").ljust(24),
          str(z).rjust(16), str(nm).rjust(8), str(nm % 8).rjust(8), "  ", img[b],
          "  square in L ?", is_sq(z))
print()
print("the true norms x^3+b are", [xP[b]^3 + b for b in [2,-2,6,-6]],
      "-- all = 1 mod 8, i.e. squares in Q_2, as N(delta) = y^2 requires")
print()
print("all four coords equal ?", len(set(img.values())) == 1, "  common value:", img[2])
print("and that class is [-5(1+pi)] =", coords(-5*(1+p0)), "= u2")
```

**Theorem.** For each $b \in \{2,-2,6,-6\}$, $V = \{\mathcal{O}, P\}$ represents
$E(\mathbb{Q}_2)/2E(\mathbb{Q}_2)$ and
$$\operatorname{Im}(\delta_b) = \big\{1,\ [-5(1+\pi)]\big\} \subset \ker N .$$

*Proof.* $\#E(\mathbb{Q}_2)/2E(\mathbb{Q}_2) = 2$ by §1, and $\delta_b(P) \neq 1$ because
$x_P - \lambda_b\pi$ is not a square in $L$ (checked above), so $P \notin 2E(\mathbb{Q}_2)$ and the two
points of $V$ represent the two cosets. $\blacksquare$

## 6. An independent brute-force scan

Sample $x \in \mathbb{Q}$ with $x^3+b$ a square in $\mathbb{Q}_2$ and record the class of
$x - \lambda_b\pi$. Exactly two classes should appear, split precisely by the sign of $v_2(x)$ — that
is the assertion $2E(\mathbb{Q}_2) = E_1$ of §1.

```python
for b in [2, -2, 6, -6]:
    seen, byk, count = {}, {}, 0
    for k in range(6):
        for a in range(-250, 251):
            if k > 0 and a % 2 == 0:
                continue
            x = QQ(a) / 4^k
            if not Q2(x^3 + b).is_square():
                continue
            count += 1
            c = coords(L0(x) - lam[b]*p0)
            seen.setdefault(c, x)
            byk.setdefault(-2*k, set()).add(c)
    print("b =", str(b).rjust(2), "  sampled", str(count).rjust(4), " classes reached:",
          sorted(seen), "  first at x =", [seen[c] for c in sorted(seen)])
    print("        by valuation:", {v: sorted(byk[v]) for v in sorted(byk, reverse=True)})
print()
print("integral x  <=>  the non-trivial class;  v_2(x) < 0  <=>  E_1 = 2E(Q_2)  <=>  trivial class")
```

## 7. Duality: the *plain* Hilbert symbol of $L$

Local Tate duality makes $\operatorname{Im}(\delta)$ maximal isotropic in $H^1(\mathbb{Q}_2,E[2])$,
here of order $\sqrt{4} = 2$ inside a group of order $4$. The pairing is the cup product followed by
corestriction from $L$; since $[L:\mathbb{Q}_2] = 3$ is odd, corestriction is the identity on
$\operatorname{Br}[2]$, so the pairing is simply
$$\langle a, b\rangle \;=\; (a,b)_L ,$$
the Hilbert symbol of $L$ — **no Frobenius twist**. (In the companion notebooks the algebra was
$\mathbb{Q}_2 \times \mathbb{Q}_4$; the twist there was exactly the contribution of the discarded
$\mathbb{Q}_2$-factor. With $L$ a field there is nothing to discard.)

Symbols are computed in $\mathbb{Q}(\sqrt[3]{2})$ at its unique prime above $2$, whose completion is $L$.

```python
PP = L0.primes_above(2)[0]
print("prime above 2 :", PP, "   e =", PP.ramification_index(), "  f =", PP.residue_class_degree())

def hs(u, v):
    return L0.hilbert_symbol(u, v, PP)

kr = [rep(m) for m in ker]
print()
print("Hilbert symbol on ker N, in the order", [str(z) for z in kr], ":")
show(matrix(4, 4, lambda i, j: hs(kr[i], kr[j])))
```

```python
print("alternating ((z,z) = 1 for all z) ?", all(hs(z, z) == 1 for z in kr))
print("non-degenerate (every non-trivial class pairs to -1 with something) ?",
      all(any(hs(kr[i], kr[j]) == -1 for j in range(4)) for i in range(1, 4)))
print()
print("=> all three subgroups of order 2 in ker N are maximal isotropic:")
for z in kr[1:]:
    ann = [w for w in kr if hs(z, w) == 1]
    print("   <" + str(z).rjust(18) + ">   annihilator =", [str(w) for w in ann],
          "  self-annihilating ?", sorted(coords(w) for w in ann) == sorted([coords(L0(1)), coords(z)]))
print()
print("Im(delta) = <-5(1+pi)> is one of them, and duality does NOT single it out.")
```

## 8. Comparison with the unit family $v_2(b) = 0$

| | $v_2(b) = 0$: $b = 1,-1,3,-3$ | $v_2(b) = 1$: $b = 2,-2,6,-6$ |
|---|---|---|
| $x^3+b$ over $\mathbb{Q}_2$ | linear $\times$ quadratic | irreducible (Eisenstein) |
| descent algebra | $\mathbb{Q}_2 \times \mathbb{Q}_4$ (unramified) | $L = \mathbb{Q}_2(\sqrt[3]2)$, ramified cubic |
| $\operatorname{Gal}$ on $E[2]$ | $\mathbb{Z}/2$ | $S_3$ |
| $E[2](\mathbb{Q}_2)$ | order $2$ | trivial |
| $\#H^1(\mathbb{Q}_2,E[2])$ | $16$ | $4$ |
| $\#E(\mathbb{Q}_2)/2E(\mathbb{Q}_2)$ | $4$ | $2$ |
| $E(\mathbb{Q}_2)$ | $\mathbb{Z}_2 \oplus \mathbb{Z}/2$ (or $\oplus\mathbb{Z}/6$) | $\mathbb{Z}_2$ |
| Tate pairing | *Frobenius-twisted* symbol of $\mathbb{Q}_4$ | *plain* symbol of $L$ |
| Lagrangians available | $15$ | $3$ |
| Lagrangians realised | $2$, Frobenius conjugates | $1$, the same for all four |

In both families the four values of $b$ exhaust the $\mathbb{Q}_2$-isomorphism classes with that
valuation, since $y^2=x^3+b \cong y^2=x^3+b'$ over $\mathbb{Q}_2$ iff $b/b' \in (\mathbb{Q}_2^\times)^6$
and on units sixth powers are squares — so $b \bmod 8$, resp. $(b/2) \bmod 8$, is the invariant.

```python
print("the four curves are pairwise non-isomorphic over Q_2 (b/b' is not a 6th power):")
for b in [2, -2, 6, -6]:
    for bb in [2, -2, 6, -6]:
        if b < bb:
            r = QQ(b)/bb
            print("   b =", str(b).rjust(2), " b' =", str(bb).rjust(2),
                  "   b/b' =", str(r).rjust(5), "  class", cls(r),
                  "  a 6th power in Q_2 ?", Q2(r).is_square())
print()
print("yet all four Kummer images coincide:", {b: img[b] for b in [2,-2,6,-6]})
print()
print("for contrast, the unit family split 2+2; see the companion notebooks.")
```

## Summary

For all four curves $y^2 = x^3 + b$ with $b \in \{2,-2,6,-6\}$:

$$V = \{\mathcal{O},\ P\}, \qquad
\operatorname{Im}(\delta_b) = \big\{1,\ [-5(1+\pi)]\big\} \subset
\ker N \subset L^\times/(L^\times)^2, \qquad \pi = \sqrt[3]{2}.$$

- $v_2(b) = 1$ is not divisible by $3$, so $x^3+b$ is Eisenstein and the descent algebra is a
  **totally ramified cubic field**, not $\mathbb{Q}_2 \times \mathbb{Q}_4$; $\operatorname{Gal}$ acts
  through $S_3$ and $E[2](\mathbb{Q}_2) = 0$;
