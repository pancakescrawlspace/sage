# The local Kummer image at $p = 2$ for $y^2 = x^3 + b$ with $b$ a unit — summary

A synthesis of the four worksheets `local-kummer-p2.ipynb` ($b=1$), `-minus` ($b=-1$),
`-plus3` ($b=3$) and `-minus3` ($b=-3$), redone here in **one uniform frame** so that the four
answers sit in a single table. Those four notebooks are left untouched; everything below is
recomputed from scratch.

Since $y^2=x^3+b \cong y^2=x^3+b'$ over $\mathbb{Q}_2$ iff $b/b' \in (\mathbb{Q}_2^\times)^6$, and on
units sixth powers are just squares (cubing is onto $\mathbb{Z}_2^\times$), the invariant is
$b \bmod 8$: these four exhaust the $\mathbb{Q}_2$-isomorphism classes with $v_2(b) = 0$.

**The uniform frame.** For $b$ odd, $-b$ is a cube in $\mathbb{Z}_2$, so $x^3+b$ has a
$\mathbb{Q}_2$-rational root and factors as (linear)$\times$(quadratic) with the quadratic generating
$\mathbb{Q}_4 = \mathbb{Q}_2(\omega)$. Thus $E[2] = \operatorname{Ind}_{\mathbb{Q}_4}^{\mathbb{Q}_2}\mu_2$,
$H^1(\mathbb{Q}_2,E[2]) \cong \mathbb{Q}_4^\times/(\mathbb{Q}_4^\times)^2$ of order $16$, and
$\#E(\mathbb{Q}_2)/2E(\mathbb{Q}_2) = 4$. Normalising the distinguished root as
$$\rho_b = \lambda_b\,\omega, \qquad \lambda_b = \sqrt[3]{-b} \in \mathbb{Z}_2^\times,
\qquad \delta_b(x,y) = x - \lambda_b\,\omega ,$$
gives $\lambda_1 = -1$, $\lambda_{-1} = 1$, $\lambda_3 = \sqrt[3]{-3} \rightsquigarrow 5$,
$\lambda_{-3} = \sqrt[3]{3} \rightsquigarrow 3$, the last two by the substitution lemma of the
$b = \pm3$ notebooks. (This is the same shape as $\lambda_b = \sqrt[3]{-b/2}$ in the companion
worksheet for $v_2(b) = 1$, where the same four values $-1, 1, 5, 3$ appear.)

> **Note on $b = 1$.** `local-kummer-p2.ipynb` used the root $\zeta_6 = -\omega^2$, which is the
> *conjugate* of the normalised $\rho_1 = -\omega$ used here. Its image as printed there is therefore
> the Frobenius conjugate of the one tabulated below. Nothing is inconsistent: §7 explains exactly
> how much of such a comparison is canonical — and the answer is less than the earlier notebooks'
> phrasing suggested.

**The table.** All four have $E(\mathbb{Q}_2)/2E(\mathbb{Q}_2) \cong (\mathbb{Z}/2)^2$, norms running
through all four square classes $1,3,7,5$ of $\mathbb{Z}_2^\times$, and

| $b$ | $\delta_b$ | $\operatorname{Im}(\delta_b)$ |
|---|---|---|
| $1$ | $x+\omega$ | $\langle[\omega-1],\ [\omega-2]\rangle$ |
| $-1$ | $x-\omega$ | $\langle[1-\omega],\ [-3-\omega]\rangle$ |
| $3$ | $x-5\omega$ | $\langle[1-\omega],\ [-3-\omega]\rangle$ |
| $-3$ | $x-3\omega$ | $\langle[\omega-1],\ [3+\omega]\rangle$ |

Rows $1$ and $-3$ name the same subgroup ($[\omega-2] = [3+\omega]$), as do rows $-1$ and $3$; each is
written with the representatives natural to its own notebook. So relative to *this* normalisation the four split $\{-1,3\}$ against $\{1,-3\}$, the two halves being
Frobenius conjugates. §6 locates the images inside the canonical structure of
$\mathbb{Q}_4^\times/(\mathbb{Q}_4^\times)^2$ — where they turn out to be pinned down completely — and §7
is about what the $\{-1,3\}$ / $\{1,-3\}$ statement does and does not mean.

## 2. $\mathbb{Q}_4^\times/(\mathbb{Q}_4^\times)^2$ and the basis

Order $2^{d+2} = 16$ for $d = [\mathbb{Q}_4:\mathbb{Q}_2] = 2$; a unit is a square as soon as it is one
mod $8$, and for $a \in \mathbb{Q}_2^\times$: $a$ is a square in $\mathbb{Q}_4$ iff $v_2(a)$ is even and
its unit part is $\equiv 1$ or $5 \pmod 8$. Basis $B = (2,\ -1,\ 1-\omega,\ -3-\omega)$, as in the
companion notebooks.

```python
K2.<w>  = Qq(4, 40)
KK.<om> = CyclotomicField(3)

def to2(z):
    a, b = z.list(); return K2(a) + K2(b)*w

def nrm(z):
    a, b = z.list(); return a^2 - a*b + b^2

def is_sq(z):
    return to2(z).is_square()

def cls(q):
    q = QQ(q); v = q.valuation(2); u = q / 2^v
    return (v % 2, ZZ(u.numerator() * ZZ(u.denominator()).inverse_mod(8)) % 8)

def sig(z):
    "Frobenius of Q_4/Q_2:  om |-> om^2"
    a, b = z.list(); return a + b*om^2

B = [KK(2), KK(-1), 1 - om, -3 - om]

def rep(m):
    z = KK(1)
    for i in range(4):
        if (m >> i) & 1: z *= B[i]
    return z

def coords(z):
    for m in range(16):
        if is_sq(z * rep(m)):
            return tuple((m >> i) & 1 for i in range(4))
    raise ValueError("not in the span of B")

print("basis (2, -1, 1-om, -3-om) independent ?", [m for m in range(16) if is_sq(rep(m))] == [0])
print("16 classes, with norms:",
      sorted(set(nrm(rep(m)) for m in range(16))))
```

## 3. The master table

Each $V$ is taken from the corresponding notebook; the $x$-coordinates that are irrational
($T = (\sqrt[3]{-b},0)$ for $b = \pm3$) are replaced by their rational substitutes, which the
substitution lemma shows changes no class.

```python
Vname = {  1: ["O", "T = (-1,0)",  "P = (-2,y)",        "P+T = (-4,-3y)"],
          -1: ["O", "T = (1,0)",   "Q = (1/4,3y/8)",    "Q+T = (-3,-2y)"],
           3: ["O", "T = (e,0)",   "P = (1,2)",         "Q = (1/4,sqrt(193)/8)"],
          -3: ["O", "T = (phi,0)", "P = (-9,2sqrt(-183))", "Q = (1/4,sqrt(-191)/8)"]}
Vx    = {  1: [None, -1, -2, -4],
          -1: [None, 1, QQ(1)/4, -3],
           3: [None, 5, 1, QQ(1)/4],
          -3: [None, 3, -9, QQ(1)/4]}

img = {}
for b in [1, -1, 3, -3]:
    print("b =", str(b).rjust(2), "    delta(x,y) = x - (" + str(lam[b]) + ")*om")
    cs = []
    for nm, x in zip(Vname[b], Vx[b]):
        z = KK(1) if x is None else KK(x) - lam[b]*om
        cs.append(coords(z))
        print("    " + nm.ljust(23) + str(z).rjust(13) + "   N =" + str(nrm(z)).rjust(8)
              + "   " + str(cls(nrm(z))) + "   " + str(coords(z)))
    img[b] = sorted(cs)
    print("    image:", img[b], "   all four norm classes ?",
          sorted(cls(nrm(KK(1) if x is None else KK(x) - lam[b]*om)) for x in Vx[b])
          == [(0,1),(0,3),(0,5),(0,7)])
    print()
print("Im(-1) == Im(3) ?", img[-1] == img[3], "    Im(1) == Im(-3) ?", img[1] == img[-3],
      "    Im(1) == Im(-1) ?", img[1] == img[-1])
```

## 4. The filtration, and where $b = 1$ is different

For $b = -1, 3, -3$ the reduction is type $\mathrm{II}$ with $c_2 = 1$, so $E(\mathbb{Q}_2) = E_1 \oplus
\langle T\rangle \cong \mathbb{Z}_2 \oplus \mathbb{Z}/2$ is pro-$2$, $2E(\mathbb{Q}_2) = E_2$, and the
norm classes label the filtration exactly:
$$E_2 \leftrightarrow 1, \qquad E_1\smallsetminus E_2 \leftrightarrow 5, \qquad
T+E_2 \leftrightarrow 3, \qquad T+(E_1\smallsetminus E_2) \leftrightarrow 7 \pmod 8 .$$
For $b = 1$ the type is $\mathrm{IV}$ with $c_2 = 3$: $E(\mathbb{Q}_2) \cong \mathbb{Z}_2 \oplus
\mathbb{Z}/6$, the extra $\mathbb{Z}/3$ sits inside $2E(\mathbb{Q}_2)$, and integral $x$ can reach the
trivial class too — e.g. $(0,\pm1)$ is $3$-torsion. The scan sees the difference at once:

```python
for b in [1, -1, 3, -3]:
    byk = {}
    for k in range(5):
        for a in range(-250, 251):
            if k > 0 and a % 2 == 0: continue
            x = QQ(a) / 4^k
            if not Q2(x^3 + b).is_square(): continue
            byk.setdefault(-2*k, set()).add(coords(KK(x) - lam[b]*om))
    print("b =", str(b).rjust(2), " classes by v_2(x):",
          {v: sorted(byk[v]) for v in sorted(byk, reverse=True)})
print()
print("b = -1, 3, -3 : integral x gives exactly two classes; v_2(x) = -2 gives a third;")
print("                v_2(x) <= -4 gives the trivial one.")
print("b = 1         : integral x already gives all four -- the Z/3 from c_2 = 3.")
```

## 5. Duality

Local Tate duality forces each image to be maximal isotropic of order $4$ in the $16$-element group.
The pairing is the **Frobenius-twisted** Hilbert symbol $(a, \sigma b)_{\mathbb{Q}_4}$ — the twist being
the contribution of the discarded $\mathbb{Q}_2$-factor of the split algebra
$\mathbb{Q}_2[x]/(x^3+b) = \mathbb{Q}_2 \times \mathbb{Q}_4$. Symbols are computed in
$\mathbb{Q}(\omega)$ at its inert prime above $2$, whose completion is $\mathbb{Q}_4$.

```python
P2 = KK.primes_above(2)[0]
def hs(u, v): return KK.hilbert_symbol(u, v, P2)

print("the twisted form is alternating:", all(hs(rep(m), sig(rep(m))) == 1 for m in range(16)))
nlag = len(set(frozenset([0, m, n, m ^^ n]) for m in range(1,16) for n in range(1,16)
               if m != n and hs(rep(m), sig(rep(n))) == 1))
print("number of Lagrangians (maximal isotropic subgroups of order 4):", nlag)
print()
for b in [1, -1, 3, -3]:
    gens = [rep(m) for m in range(16) if tuple((m >> i) & 1 for i in range(4)) in img[b]
            and m != 0][:2]
    ann = sorted(coords(rep(m)) for m in range(16)
                 if all(hs(rep(m), sig(g)) == 1 for g in gens))
    print("b =", str(b).rjust(2), "  image is isotropic ?",
          all(hs(g, sig(h)) == 1 for g in gens for h in gens),
          "   self-annihilating ?", ann == img[b])
```

## 6. Where the image sits inside $\mathbb{Q}_4^\times/(\mathbb{Q}_4^\times)^2$

The target carries structure of its own, and asking how the image meets it gives statements that
survive the ambiguity of §7 — because everything below is $\sigma$-stable.

A word of care first: the familiar description
$$\mathbb{Q}_4^\times \cong 2^{\mathbb{Z}} \times \mu_3 \times \{\pm1\} \times \mathbb{Z}_2^2$$
is **not** canonical as a direct product. Splitting off $2^{\mathbb{Z}}$ needs a uniformiser, and
splitting the $\mathbb{Z}_2^2$ off the principal units needs a further choice. What *is* canonical is a
filtration together with two extra pieces of structure:
$$1 \;\subset\; U^{(2)}(\mathbb{Q}_4^\times)^2/(\mathbb{Q}_4^\times)^2 \;\subset\;
U^{(1)}(\mathbb{Q}_4^\times)^2/(\mathbb{Q}_4^\times)^2 =: V_1 \;\subset\; \mathbb{Q}_4^\times/(\mathbb{Q}_4^\times)^2,$$
of orders $1 \subset 2 \subset 8 \subset 16$, plus the subgroup $\mu_2 = \{1,[-1]\}$ and the norm $N$ to
$\mathbb{Q}_2^\times/(\mathbb{Q}_2^\times)^2$. Three identifications come out of the computation below:

- $V_1$ is exactly the **even-valuation** part, i.e. the classes of units (the odd part $\mu_3$ dies mod
  squares, so unit classes and $U^{(1)}$-classes agree);
- $U^{(2)}(\mathbb{Q}_4^\times)^2/(\mathbb{Q}_4^\times)^2$ is exactly the **unramified line**
  $\{[a] : \mathbb{Q}_4(\sqrt a)/\mathbb{Q}_4 \text{ unramified}\}$, of order $2$;
- $\ker\big(N|_{V_1}\big) = \{1,[-1]\}$ precisely, so $N$ induces a canonical isomorphism
  $$V_1/\{1,[-1]\} \;\xrightarrow{\ \sim\ }\; \mathbb{Z}_2^\times/(\mathbb{Z}_2^\times)^2 .$$
  That quotient is the "third factor"; canonically it is a quotient, not a summand.

```python
mof = {coords(rep(m)): m for m in range(16)}      # class -> index
V1  = [m for m in range(16) if to2(rep(m)).valuation() % 2 == 0]
m_1 = mof[coords(KK(-1))]

print("V1 = even-valuation classes :", len(V1), "of 16      [-1] in V1 ?", m_1 in V1)
print()
print("higher unit filtration, mod squares:")
for mm in [1, 2, 3]:
    S = set(coords(KK(1) + 2^mm*(KK(a) + c*om)) for a in range(8) for c in range(8))
    print("   U^(" + str(mm) + ")  ->", len(S), "classes ", sorted(S))
print()
ur = [m for m in range(1, 16)
      if all((hs(rep(n), rep(m)) == 1) == (to2(rep(n)).valuation() % 2 == 0) for n in range(16))]
print("unramified line { [a] : Q_4(sqrt a)/Q_4 unramified } :",
      [(coords(rep(m)), str(rep(m)), nrm(rep(m))) for m in ur])
print()
print("the norm on V1:")
for m in V1:
    print("   " + str(coords(rep(m))) + str(rep(m)).rjust(11) + "   N =" + str(nrm(rep(m))).rjust(4)
          + "   class " + str(cls(nrm(rep(m)))))
print("ker(N | V1) =", [coords(rep(m)) for m in V1 if cls(nrm(rep(m))) == (0,1)],
      "  = {1, [-1]} ?", sorted(m for m in V1 if cls(nrm(rep(m))) == (0,1)) == sorted([0, m_1]))
print("N(V1) =", sorted(set(cls(nrm(rep(m))) for m in V1)), " = all of Z_2^*/(Z_2^*)^2")
```

**Proposition.** For every unit $b$, writing $\delta_b(x,y) = x - \lambda_b\omega$ with
$\lambda_b \in \mathbb{Z}_2^\times$:

1. $\operatorname{Im}(\delta_b) \subseteq V_1$;
2. $\operatorname{Im}(\delta_b) \cap \{1,[-1]\} = 1$;
3. $N$ carries $\operatorname{Im}(\delta_b)$ isomorphically onto $\mathbb{Z}_2^\times/(\mathbb{Z}_2^\times)^2$;

equivalently, $\operatorname{Im}(\delta_b)$ is a **complement to $\mu_2$ inside $V_1$** — a section of
$1 \to \{1,[-1]\} \to V_1 \xrightarrow{N} \mathbb{Z}_2^\times/(\mathbb{Z}_2^\times)^2 \to 1$. It also
contains the unramified line.

*Proof of 1.* $\{1,\omega\}$ is an integral basis, so $v(\alpha+\beta\omega) = \min(v\alpha, v\beta)$;
the $\omega$-coefficient of $\delta_b$ is the unit $-\lambda_b$, whence $v(\delta_b) = \min(v_2(x), 0)$.
If $v_2(x) \geq 0$ this is $0$. If $v_2(x) < 0$ then $2v_2(y) = v_2(y^2) = v_2(x^3+b) = 3v_2(x)$, so
$v_2(x)$ is even. Either way $v(\delta_b)$ is even. $\square$

Given 1, statements 2 and 3 are the same, since $\ker(N|_{V_1}) = \{1,[-1]\}$; and 3 is the
"norms run through all four square classes" observation of §3, in invariant form. Note the proof of 1
uses only that *negative* valuations of $x$ are even — for $b = 1$ the point $(2,3)$ has $v_2(x) = 1$.

```python
print("v_2(x) occurring on E(Q_2), by curve:")
for b in [1, -1, 3, -3]:
    vs = set()
    for k in range(6):
        for a in range(-200, 201):
            if k > 0 and a % 2 == 0: continue
            x = QQ(a) / 4^k
            if Q2(x^3 + b).is_square(): vs.add(x.valuation(2))
    print("   b =", str(b).rjust(2), sorted(vs), "   negative ones all even ?",
          all(v % 2 == 0 for v in vs if v < 0))
print()
print(" b    Im c V1 ?   [-1] in Im ?   N|Im injective ?   contains the unramified line ?")
for b in [1, -1, 3, -3]:
    I = img[b]
    print(str(b).rjust(2), "  ",
          str(all(mof[c] in V1 for c in I)).ljust(11),
          str(coords(KK(-1)) in I).ljust(14),
          str(len(set(cls(nrm(rep(mof[c]))) for c in I)) == 4).ljust(18),
          coords(rep(ur[0])) in I)
```

### How much this pins down

Being a complement to $\mu_2$ in $V_1$ is already restrictive; adding maximal isotropy for the
twisted pairing leaves exactly two subgroups — and they are a single Frobenius orbit. So the position
of the image in the canonical structure is completely determined, which is *why* the four curves could
not have differed at the invariant level of §7.

```python
V1s = set(V1)
lags = set()
for m in range(1, 16):
    for n in range(1, 16):
        if m != n and hs(rep(m), sig(rep(n))) == 1:
            lags.add(frozenset([0, m, n, m ^^ n]))
lags = [L for L in lags if len(L) == 4]

subs = set()
for m in V1:
    for n in V1:
        if m and n and m != n:
            S = frozenset([0, m, n, m ^^ n])
            if len(S) == 4 and S <= V1s: subs.add(S)

good = [L for L in lags if L <= V1s and m_1 not in L]
print("order-4 subgroups of V1                        :", len(subs))
print("   ... complements to {1,[-1]} (no duality)    :", len([S for S in subs if m_1 not in S]))
print()
print("Lagrangians for the twisted pairing            :", len(lags))
print("   ... contained in V1                         :", len([L for L in lags if L <= V1s]))
print("   ... and avoiding [-1]                       :", len(good))
print()
for L in good:
    print("   ", sorted(coords(rep(m)) for m in L))
print("these two are exactly the realised images ?",
      sorted([sorted(coords(rep(m)) for m in L) for L in good]) == sorted([img[1], img[-1]]))
print("and they form a single Frobenius orbit ?",
      sorted(sorted(coords(sig(rep(m))) for m in L) for L in good)
      == sorted([sorted(coords(rep(m)) for m in L) for L in good]))
```

## 7. How canonical is the comparison? — a correction

The table of §3 compares four *different* curves, so it presupposes an identification of their
$2$-torsion modules. That identification is **not** unique, and the ambiguity is exactly the Frobenius.
Concretely: if $M, M'$ are isomorphic $G$-modules, the set of $G$-isomorphisms $M \to M'$ is a torsor
under $\operatorname{Aut}_G(M)$. Here $E[2] \cong \mathbb{F}_2^2$ with
$\operatorname{Aut}(\mathbb{F}_2^2) = GL_2(\mathbb{F}_2) \cong S_3$ permuting the three non-zero
vectors — i.e. the three roots of $x^3+b$ — and $G$ acts through the subgroup generated by the
transposition $\tau$ swapping the two conjugate roots. So
$$\operatorname{Aut}_G(E[2]) = C_{S_3}(\tau) = \{1, \tau\} \cong \mathbb{Z}/2 ,$$
and $\tau$ — the swap — is precisely what changes the choice of distinguished root, hence acts on
$H^1(\mathbb{Q}_2,E[2]) \cong \mathbb{Q}_4^\times/(\mathbb{Q}_4^\times)^2$ as $\sigma$. Two consequences:

- **Relative to an arbitrary $G$-module identification**, only the *unordered pair*
  $\{\operatorname{Im}(\delta_b),\ \sigma\operatorname{Im}(\delta_b)\}$ is well defined. None of the four
  images is $\sigma$-stable, and all four of these pairs turn out to be **the same**. At this level the
  four curves are indistinguishable, and the "two Lagrangians exchanged by Frobenius" reading of the
  earlier notebooks says nothing: the same ambiguity would even make a curve fail to be comparable
  with *itself*.
- **Relative to the geometric identification** there is no ambiguity. Any isomorphism
  $E_b \to E_{b'}$ over $\overline{\mathbb{Q}}_2$ is $(x,y)\mapsto(u^2x,u^3y)$ with $u^6 = b'/b$, and
  induces $x \mapsto \nu x$ on $2$-torsion with $\nu^3 = b'/b$; of the three such $\nu$ exactly one lies
  in $\mathbb{Q}_2$ (units are cubes, $\omega \notin \mathbb{Q}_2$), and only that one is Galois
  equivariant. That unique map is the normalisation $\rho_b = \sqrt[3]{-b}\,\omega$ used above, and
  relative to it the split $\{-1,3\}$ / $\{1,-3\}$ is a genuine statement.

The two notions really do differ: $\operatorname{Aut}(E) = \mu_6$ acts on $E[2]$ through $\mu_3$, i.e.
by $3$-**cycles** of the roots, whereas $\tau$ is a **transposition** — so the swap is a Galois-module
automorphism that comes from no automorphism of the curve.

```python
S3 = SymmetricGroup(3)
tau, cyc = S3([(2,3)]), S3([(1,2,3)])

for nm, G in [("unit family  (Galois image = <tau>, order 2)", S3.subgroup([tau])),
              ("even family  (Galois image = S_3)",            S3)]:
    C = [g for g in S3 if all(g*h == h*g for h in G)]
    print(nm.ljust(44), " #Aut_G(E[2]) =", len(C), " ", [str(g) for g in C])
print()
print("Aut(E) = mu_6 acts on E[2] through mu_3, by 3-cycles:", [str(cyc), str(cyc^2)])
print("tau is a transposition, so it is NOT induced by any automorphism of the curve:",
      tau not in [S3.one(), cyc, cyc^2])
print()
print("the three cube roots of b'/b: exactly one is in Q_2, so exactly one of the three")
print("geometric identifications E_b[2] -> E_b'[2] is Galois equivariant.")
for bb in [-1, 3, -3]:
    print("   b = 1, b' =", str(bb).rjust(2), "  b'/b =", str(bb).rjust(2),
          "  nu_0 = cbrt(b'/b) =", Z2(bb).nth_root(3).lift() % 128, "mod 128,",
          " the other two are nu_0*om, nu_0*om^2, not in Q_2")
```

```python
print("the invariant object: the sigma-orbit of the image")
print()
orb = {}
for b in [1, -1, 3, -3]:
    sig_img = sorted(coords(sig(rep(m))) for m in range(16)
                     if tuple((m >> i) & 1 for i in range(4)) in img[b])
    orb[b] = sorted([img[b], sig_img])
    print("b =", str(b).rjust(2), "  Im =", img[b])
    print("        sigma Im =", sig_img, "   sigma-stable ?", img[b] == sig_img)
print()
print("all four unordered pairs {Im, sigma Im} coincide ?",
      len(set(str(orb[b]) for b in [1,-1,3,-3])) == 1)
print()
print("=> up to the unavoidable ambiguity, all four unit curves give the SAME answer;")
print("   the 2+2 split is a statement about the geometric normalisation only.")
```

## 8. Contrast with $v_2(b) = 1$

## Summary

- The four unit curves $b = 1,-1,3,-3$ exhaust the $\mathbb{Q}_2$-isomorphism classes of
  $y^2 = x^3+b$ with $b \in \mathbb{Z}_2^\times$, the invariant being $b \bmod 8$;
- each has $E(\mathbb{Q}_2)/2E(\mathbb{Q}_2) \cong (\mathbb{Z}/2)^2$ with the four $\delta$-values
  separated by their norms, which run through all four square classes of $\mathbb{Z}_2^\times$;
- three of them ($b = -1, 3, -3$) are type $\mathrm{II}$ with $c_2 = 1$ and have the clean filtration
  dictionary $E_2, E_1\smallsetminus E_2, T+E_2, T+(E_1\smallsetminus E_2) \leftrightarrow 1,5,3,7$;
  $b = 1$ is type $\mathrm{IV}$ with $c_2 = 3$ and behaves differently, integral $x$ already meeting
  all four classes;
- every image is maximal isotropic for the Frobenius-twisted Hilbert symbol of $\mathbb{Q}_4$;
- **position in the canonical structure (§6):** each image lies in the even-valuation part $V_1$, meets
  $\{1,[-1]\}$ trivially, and is carried by the norm isomorphically onto
  $\mathbb{Z}_2^\times/(\mathbb{Z}_2^\times)^2$ — i.e. it is a complement to $\mu_2$ inside $V_1$, and it
  contains the unramified line. Together with maximal isotropy this leaves exactly two subgroups,
  forming one Frobenius orbit, so the image is determined up to precisely the ambiguity of §7;
- **on comparison:** identifications of the four $E_b[2]$ form a torsor under
  $\operatorname{Aut}_G(E[2]) \cong \mathbb{Z}/2$, generated by the swap of the two conjugate
  $2$-torsion points, which acts as Frobenius on $\mathbb{Q}_4^\times/(\mathbb{Q}_4^\times)^2$. Relative
  to the unique *geometric* identification the four split $\{-1,3\}$ / $\{1,-3\}$ into Frobenius
  conjugates; relative to abstract $G$-module identifications only the pair
  $\{\operatorname{Im}, \sigma\operatorname{Im}\}$ is defined, and then all four agree. The earlier
  notebooks' phrasing ("the twist carries the image to a different Lagrangian") is true only in the
  first sense and should be read with that qualifier.
