# Does Mizukami's isomorphism descend? The Galois twist of the Fermat quartic

The companion notebook [fermat-quartic-kummer.ipynb](fermat-quartic-kummer.ipynb) verifies Mizukami's
isomorphism $\psi: X \xrightarrow{\sim} K = \operatorname{Kum}(A)$ between the Fermat quartic and the
Kummer surface of $A \cong E_i \times E_{2i}$, defined over $k = \mathbb{Q}(\mu_8)$ and not over
$\mathbb{Q}$.

This notebook asks what that costs. If one wants to describe $k_0$-points of the quartic for a subfield
$k_0 \subsetneq k$ — $\mathbb{Q}$ or $\mathbb{Q}(i)$ — via the Kummer surface, one has to carry the
$\mathbb{Q}$-structure of $X$ across $\psi$, and it arrives on $K$ as a *twisted* Galois action. Two
questions:

1. What is that twisted action, explicitly?
2. Can it be made to live on the abelian surface $A$ — so that the classical dictionary
   "points on a Kummer surface $=$ points on twists of $A$, modulo $\pm 1$" survives the descent?

The answer to (1) is: clean for $\sigma_5$, resistant for $\sigma_7$. The answer to (2), for
$k_0 = \mathbb{Q}$, is **no**, and sections 5–8 prove it as far as a finite computation can: the
obstruction is a lattice-theoretic fact about $\operatorname{Pic}(\bar X)$ as a Galois module, not an
artifact of a bad choice of $\psi$. (For $k_0 = \mathbb{Q}(\sqrt2)$ or $\mathbb{Q}(\sqrt{-2})$ the
answer turns out to be yes; that is the subject of the companion notebook
[fermat-quartic-kummer-fields.ipynb](fermat-quartic-kummer-fields.ipynb).)

## The formalism

$X$ and $K$ are both defined over $\mathbb{Q}$ — the equations (6) of $K$ have rational coefficients —
and $\psi$ is defined over $k$. So $X$ is a $\mathbb{Q}$-form of $K$, classified by the cocycle
$$c_\sigma \;=\; {}^\sigma\!\psi \circ \psi^{-1} \;\in\; \operatorname{Aut}_k(K), \qquad
\sigma \in \Gamma = \operatorname{Gal}(k/\mathbb{Q}) = \{1,\sigma_3,\sigma_5,\sigma_7\} \cong (\mathbb{Z}/2)^2,$$
where $\sigma_a(\zeta_8) = \zeta_8^a$. For $P \in X(\bar{\mathbb{Q}})$ with $Q = \psi(P)$, the point $P$
is $k_0$-rational precisely when
$$ {}^\sigma Q \;=\; c_\sigma(Q) \qquad \text{for all } \sigma \in \operatorname{Gal}(k/k_0). $$
The cocycle is well defined only up to coboundary: replacing $\psi$ by $\alpha\psi$, $\alpha \in
\operatorname{Aut}_k(K)$, replaces $c_\sigma$ by $\alpha\, c_\sigma\, {}^\sigma\!\alpha^{-1}$. That
freedom is the whole point of sections 5–8.

Throughout, the coordinates on $K$ are the ones from the companion notebook, but repackaged: instead of
$(w_1,w_2,y,z)$ use $(\Theta, y, z)$ with $w_1 = -\Theta/z$, $w_2 = -\Theta z$. This is what makes
$\sigma_5$ come out as a Möbius transformation.

## 1. The cocycle at $\sigma_5$: descent to $\mathbb{Q}(i)$

$\sigma_5$ fixes $i$ and negates $\sqrt2$, and it generates $\operatorname{Gal}(k/\mathbb{Q}(i))$. Since
$e'_+, e''_+, h'_{13}, h''_{11}, h''_{33}$ involve only $i$, they are fixed, so
$$z = -\sqrt2\,\frac{e'_+}{h'_{13}} \longmapsto -z, \qquad y = 2i\,\frac{e''_+}{h''_{11}} \longmapsto y$$
are immediate. For $\Theta$: because $w_1w_2 = \Theta^2$ and $w_2/w_1 = z^2$, both $w_j$ must be scaled
by one and the same factor $\rho$, and preserving
$y^2 = \frac{(w_1-1)(w_2-1)}{(w_1-\frac12)(w_2-\frac12)}$ makes $\rho$ a root of a quadratic whose other
root is $1$. Unwinding, with $t := z + z^{-1}$,
$$\boxed{\;c_{\sigma_5}:\quad \Theta \;\longmapsto\; \frac{3\Theta+t}{2t\,\Theta+3},\qquad
y \longmapsto y, \qquad z \longmapsto -z. \;}$$
Since $z \mapsto -z$ sends $t \mapsto -t$, composing twice multiplies by $9-2t^2$ — the identity, as the
cocycle relation ${}^\sigma c_\sigma = c_\sigma^{-1}$ demands.

Verification below is over $\mathbb{F}_{p^2}$ for $p \equiv 5 \bmod 8$, where $\zeta_8 \in
\mathbb{F}_{p^2}$ and the $p$-power Frobenius acts exactly as $\sigma_5$. Using
$c(\psi(P)) = \bigl(\psi(P^{(p)})\bigr)^{(p)}$, the formula can be tested pointwise.

```python
def setup(p):
    Fq = GF(p^2,'a'); g = Fq.multiplicative_generator(); z8 = g^((p^2-1)//8)
    assert z8^4 == -1
    return Fq, z8

def psiT(P, z8):
    "Mizukami's map in the coordinates (Theta, y, z);  w1 = -Theta/z, w2 = -Theta*z"
    x0,x1,x2,x3 = P; iu = z8^2; r2 = z8 + z8^7
    epP, epPP = x0*x3+x1*x2, x0*x2+x1*x3
    h13 = x0^2 - iu*x1^2 + iu*x2^2 +    x3^2
    h11 = x0^2 - iu*x1^2 -    x2^2 - iu*x3^2
    h33 = x0^2 + iu*x1^2 -    x2^2 + iu*x3^2
    N1  = (x0+z8^3*x1)^2 + (x2-z8*x3)^2 ; D1 = (x0+z8*x1)^2 + (x2+z8^3*x3)^2
    N2  = x2^2 + iu*r2*x2*x3 - x3^2     ; D2 = x0^2 - r2*x0*x1 + x1^2
    den = D1*h33*D2
    if den==0 or h13==0 or h11==0 or epP==0: return None
    Th = N1*epPP*N2/den ; z = -r2*epP/h13 ; y = 2*iu*epPP/h11
    return None if z==0 else (Th, y, z)

def c5(T):
    Th, y, z = T ; t = z + 1/z ; d = 2*t*Th + 3
    return None if d==0 else ((3*Th+t)/d, y, -z)

def sample_points(Fq, N):
    out = []
    for _ in range(N):
        x0,x1,x2 = (Fq.random_element() for _ in range(3))
        c = -(x0^4+x1^4+x2^4)
        if c == 0: continue
        try: x3 = c.nth_root(4)
        except ValueError: continue
        out.append((x0,x1,x2,x3))
    return out

print("Frobenius = sigma_5 :  Theta -> (3*Theta+t)/(2t*Theta+3),  y -> y,  z -> -z,   t = z + 1/z")
for p in [13, 29, 37, 53]:
    Fq, z8 = setup(p) ; ok = tot = inv = 0
    for P in sample_points(Fq, 2500):
        T = psiT(P, z8) ; Tf = psiT(tuple(u^p for u in P), z8)
        if T is None or Tf is None: continue
        R = tuple(u^p for u in Tf) ; tot += 1
        pr = c5(T)
        if pr is not None and pr == R: ok += 1
        bk = c5(pr) if pr is not None else None
        if bk is not None and bk == T: inv += 1
    print("   p = %2d :  formula holds %4d/%4d      c^2 = id on %4d/%4d" % (p, ok, tot, inv, tot))
print()
print("(the few misses are points where a denominator vanishes)")
```

Writing $\Theta = \alpha + \beta\sqrt2$ and $z = \sqrt2\,s$ with $\alpha,\beta,s \in \mathbb{Q}(i)$, and
setting $\tilde t = s + \frac1{2s}$ and $N = N_{k/\mathbb{Q}(i)}(\Theta) = \alpha^2 - 2\beta^2$, the
condition ${}^{\sigma_5}Q = c_{\sigma_5}(Q)$ becomes
$$y \in \mathbb{Q}(i), \qquad z \in \sqrt2\,\mathbb{Q}(i), \qquad \tilde t\,(2N-1) = 6\beta .$$
One norm equation. That is the description of $X(\mathbb{Q}(i))$ — for the **Fermat** quartic. Note that
over $\mathbb{Q}(i)$ the diagonal quartic $x^4+y^4+z^4=w^4$ is *not* isomorphic to it: the change of
variable needs $u^4=-1$, and no such $u$ lies in $\mathbb{Q}(i)$.

## 2. The cocycle at $\sigma_7$ resists

Complex conjugation is a different story. Every element of $k(K)$ is of the form
$\frac{P_0+P_1y+P_2z+P_3yz}{Q_0+Q_1y+Q_2z+Q_3yz}$ with $P_i,Q_i \in k(w_1,w_2)$, so $c_{\sigma_7}$ *has*
such an expression; the question is how big. Fitting by linear algebra over $\mathbb{F}_{p^2}$ with
$p \equiv 7 \bmod 8$, with an ansatz of bidegree $\le d$ in $(w_1,w_2)$:

```python
def psiW(P, z8):
    "the same map in the coordinates (w1, w2, y, z)"
    T = psiT(P, z8)
    if T is None: return None
    Th, y, z = T
    return (-Th/z, -Th*z, y, z)

def monos(d): return [(i,j,k,l) for i in range(d+1) for j in range(d+1) for k in (0,1) for l in (0,1)]
def ev(m,Q): return Q[0]^m[0]*Q[1]^m[1]*Q[2]^m[2]*Q[3]^m[3]

def try_fit(data, slot, dmax, Fq, lab):
    pts = [(Q, R[slot]) for (Q,R) in data]
    for d in range(dmax+1):
        M = monos(d); nm = len(M); need = 2*nm + 10
        if len(pts) < need:
            print("      %-3s: too few points for bidegree %d" % (lab,d)); return
        rows = [[-ev(m,Q) for m in M] + [v*ev(m,Q) for m in M] for (Q,v) in pts[:need]]
        ker = matrix(Fq, rows).right_kernel()
        if ker.dimension() == 0: continue
        for b in ker.basis():
            num = {M[t]: b[t]    for t in range(nm) if b[t]!=0}
            den = {M[t]: b[nm+t] for t in range(nm) if b[nm+t]!=0}
            if not den: continue
            good = 0
            for (Q,v) in pts[:400]:
                de = sum(c*ev(m,Q) for m,c in den.items())
                if de != 0 and sum(c*ev(m,Q) for m,c in num.items())/de == v: good += 1
            if good > 0.85*min(400,len(pts)):
                print("      %-3s: FIT at bidegree %d  (%d/%d checked, %d num / %d den terms)"
                      % (lab, d, good, min(400,len(pts)), len(num), len(den))); return
    print("      %-3s: NO fit up to bidegree %d" % (lab, dmax))

p = 23; Fq, z8 = setup(p)          # 23 = 7 mod 8, so Frobenius acts as sigma_7
data = []
for P in sample_points(Fq, 2500):
    Q = psiW(P, z8) ; Qf = psiW(tuple(u^p for u in P), z8)
    if Q is None or Qf is None or Q[2]==0: continue
    data.append((Q, tuple(u^p for u in Qf)))
print("p = 23, Frobenius = sigma_7 ;  usable points:", len(data))
for slot, lab in [(3,'z'), (0,'w1')]:
    try_fit(data, slot, 5, Fq, lab)
```

So $c_{\sigma_7}$ needs bidegree $\ge 5$ for $z$ — with essentially every monomial appearing — and does
not fit at all for $w_1$ within the range tested (bidegree $6$ was also checked separately, likewise
nothing). The same happens for $\sigma_3 = \sigma_5\sigma_7$, and for the extra automorphism relating the
diagonal quartic to the Fermat one: even its simplest piece, $x_3 \mapsto -x_3$, has no simple action on
$(\Theta,y,z)$.

This is not just inconvenience. The next section identifies what is actually going on.

## 3. Why: the contracted 16 curves are not Galois-stable

$\psi$ contracts 16 mutually skew rational curves on $X$ to the 16 exceptional lines of
$\operatorname{Kum}(A)$. Swinnerton-Dyer's choice is
$$M_{51},M_{33},M_{15},M_{77},\ [L_{33}M_{11}],[L_{33}M_{55}],[L_{15}M_{37}],[L_{15}M_{73}],\
N_{11},N_{37},N_{55},N_{73},\ [L_{57}N_{15}],[L_{57}N_{51}],[L_{71}N_{33}],[L_{71}N_{77}],$$
in the notation $L_{\mu\nu}: x_0 = \epsilon^\mu x_1, x_2 = \epsilon^\nu x_3$ (and $M$, $N$ for the two
other pairings), $[\Lambda\Lambda']$ the residual conic in the plane of two meeting lines. Galois acts by
multiplying all indices by $a$ mod $8$.

If $c_\sigma$ preserved the 16 exceptional curves of $K$ it would be an *affine* automorphism — induced
by $a \mapsto \gamma(a) + \tau$ on $A$ — and the descent condition would translate into a condition on
points of $A$. But $c_\sigma$ maps the exceptional set to $\psi(\sigma^{-1}S)$, so it preserves it only
if $\sigma(S) = S$:

```python
S = [('M',5,1),('M',3,3),('M',1,5),('M',7,7),
     ('LM',3,3,1,1),('LM',3,3,5,5),('LM',1,5,3,7),('LM',1,5,7,3),
     ('N',1,1),('N',3,7),('N',5,5),('N',7,3),
     ('LN',5,7,1,5),('LN',5,7,5,1),('LN',7,1,3,3),('LN',7,1,7,7)]
Sset = set(S)
def act(a, cur): return (cur[0],) + tuple((a*t) % 8 for t in cur[1:])
for a in (3,5,7):
    img = [act(a,c) for c in S]
    print("   sigma_%d : %2d of the 16 curves stay in the set" % (a, sum(1 for c in img if c in Sset)))
print()
print("Under sigma_5 the eight lines survive and the eight conics are thrown out;")
print("under sigma_3 and sigma_7 nothing survives.  So no c_sigma is affine, and the")
print("descent condition cannot be phrased on the abelian surface.")
```

## 4. The real question

Because the cocycle is defined only up to coboundary, none of this is final. Choosing a different
$\psi$ contracts a different 16-set. So everything turns on:

> **Does $X$ carry a Galois-stable Kummer structure?** That is, 16 pairwise disjoint $(-2)$-curves,
> permuted as a set by $\operatorname{Gal}(\mathbb{Q}(\mu_8)/\mathbb{Q})$, whose sum is $2$-divisible in
> $\operatorname{Pic}$.

If yes, one can choose $\psi$ contracting *that* set, every $c_\sigma$ becomes affine, $X \cong
\operatorname{Kum}(V)$ over $\mathbb{Q}$ for a $2$-covering $V$ of $A$, and $\mathbb{Q}$-points of $X$
off the exceptional locus become $\pm$-orbits of $\mathbb{Q}$-points on twists of $V$ — the description
one wants. If no, the obstruction is genuine.

This is a finite question about $\operatorname{Pic}(\bar X)$ as a Galois module, and the rest of the
notebook settles it as far as computation can reach.

## 5. The Picard lattice from the 48 lines

Pinch and Swinnerton-Dyer proved $\operatorname{Pic}(\bar X) \cong \mathbb{Z}^{20}$ is generated by the
48 lines. Rather than assume the combinatorics, the intersection numbers are computed from the
coordinates: two distinct lines on a smooth quartic meet transversally or not at all, so
$\Lambda \cdot \Lambda' \in \{0,1\}$ according to whether the $4\times4$ matrix of their spanning vectors
is singular.

```python
K.<ep> = CyclotomicField(8)
odds = [1,3,5,7]
lines = []
for mu in odds:
    for nu in odds: lines.append(('L',mu,nu, matrix(K,[[ep^mu,1,0,0],[0,0,ep^nu,1]])))
for mu in odds:
    for nu in odds: lines.append(('M',mu,nu, matrix(K,[[ep^mu,0,1,0],[0,ep^nu,0,1]])))
for mu in odds:
    for nu in odds: lines.append(('N',mu,nu, matrix(K,[[ep^mu,0,0,1],[0,ep^nu,1,0]])))
labels = [(f,mu,nu) for (f,mu,nu,_) in lines] ; n = 48

Rst.<s,t> = PolynomialRing(K)
bad = sum(1 for (f,mu,nu,Bm) in lines
          if sum((s*Bm[0][j] + t*Bm[1][j])^4 for j in range(4)) != 0)
print("lines not lying on x0^4+x1^4+x2^4+x3^4 = 0 :", bad)

Gram = matrix(ZZ, n, n)
for a in range(n):
    for b in range(n):
        Gram[a,b] = -2 if a==b else (1 if lines[a][3].stack(lines[b][3]).rank() < 4 else 0)

def sdrule(A,Bl):
    f,al,be = A[:3]; g,ga,de = Bl[:3]
    if (f,al,be)==(g,ga,de): return -2
    if f==g: return 1 if ((al==ga) != (be==de)) else 0
    key = (f,g)
    if key not in [('L','M'),('L','N'),('M','N')]: key = (g,f); al,be,ga,de = ga,de,al,be
    if key==('L','M'): return 1 if (al-be-ga+de) % 8 == 0 else 0
    if key==('L','N'): return 1 if (al+be-ga+de) % 8 == 0 else 0
    return 1 if (al+be-ga-de) % 8 == 0 else 0
print("entries where the geometry disagrees with Swinnerton-Dyer's stated rules:",
      sum(1 for a in range(n) for b in range(n) if Gram[a,b] != sdrule(lines[a],lines[b])))
print("rank of the span of the 48 lines:", Gram.rank())
```

```python
bas = []
for a in range(n):
    if Gram[bas+[a], bas+[a]].rank() == len(bas)+1: bas.append(a)
GB = Gram[bas,bas] ; GBinv = GB.inverse()
linecoord = [GBinv*vector(QQ,[Gram[a,b] for b in bas]) for a in range(n)]
def pr(u,v): return u*GB*v
assert all(pr(linecoord[a],linecoord[b]) == Gram[a,b] for a in range(n) for b in range(n))
Pic = matrix(QQ, linecoord).row_module(ZZ)
Bm  = matrix(QQ, Pic.basis())
print("Pic = Z-span of the 48 lines, rank", Pic.rank(),
      ", discriminant", (Bm*GB*Bm.transpose()).det(), " (= -|disc T(X)|, so this is all of Pic)")
idx = {labels[a]: a for a in range(n)}
H = sum(linecoord[idx[('L',1,nu)]] for nu in odds)
print("H = sum_nu L_{1,nu} :  H^2 =", pr(H,H), ", H.L =", set(pr(H,linecoord[a]) for a in range(n)))
GP = matrix(ZZ, Bm*GB*Bm.transpose())              # Gram in Pic's own basis; Pic = Z^20
hG = Bm.solve_left(matrix(QQ,[H]))[0]
def ip(u,v): return u*GP*v
```

Validation: Swinnerton-Dyer's 16 curves really are a Kummer structure. The conic classes are
$[\Lambda\Lambda'] = H - \Lambda - \Lambda'$.

```python
def toPic(v): return Bm.solve_left(matrix(QQ,[v]))[0]
lc = [vector(QQ, toPic(vector(QQ,linecoord[a]))) for a in range(n)]
hv = vector(QQ, hG)
conics = {}
for b in range(n):
    for c in range(b+1,n):
        if labels[b][0] != labels[c][0] and Gram[b,c] == 1:
            conics[(labels[b],labels[c])] = hv - lc[b] - lc[c]
print("conics [Lambda Lambda'] :", len(conics),
      "; all (-2):", all(ip(v,v)==-2 for v in conics.values()),
      "; all of degree 2:", all(ip(v,hv)==2 for v in conics.values()))
def cn(A,Bl): return conics[(A,Bl)] if (A,Bl) in conics else conics[(Bl,A)]
SD = [lc[idx[('M',5,1)]], lc[idx[('M',3,3)]], lc[idx[('M',1,5)]], lc[idx[('M',7,7)]],
      cn(('L',3,3),('M',1,1)), cn(('L',3,3),('M',5,5)), cn(('L',1,5),('M',3,7)), cn(('L',1,5),('M',7,3)),
      lc[idx[('N',1,1)]], lc[idx[('N',3,7)]], lc[idx[('N',5,5)]], lc[idx[('N',7,3)]],
      cn(('L',5,7),('N',1,5)), cn(('L',5,7),('N',5,1)), cn(('L',7,1),('N',3,3)), cn(('L',7,1),('N',7,7))]
print("Swinnerton-Dyer's 16 curves: all (-2):", all(ip(v,v)==-2 for v in SD),
      "; pairwise disjoint:", all(ip(SD[i],SD[j])==0 for i in range(16) for j in range(i+1,16)),
      "; half-sum in Pic:", all(c % 2 == 0 for c in sum(SD)))
```

## 6. A rank count that forces the shape of any stable configuration

The lines are defined over $\mathbb{Q}(\mu_8)$ and generate $\operatorname{Pic}$, so $\Gamma_\mathbb{Q}$
acts through $G = \operatorname{Gal}(\mathbb{Q}(\mu_8)/\mathbb{Q}) \cong (\mathbb{Z}/2)^2$, permuting the
48 lines by $\sigma_a: L_{\mu\nu} \mapsto L_{a\mu,a\nu}$.

```python
def perm(a): return [idx[(f,(a*mu)%8,(a*nu)%8)] for (f,mu,nu) in labels]
P3,P5,P7 = perm(3),perm(5),perm(7)
seen=set(); orbsum=[]
for a in range(n):
    if a in seen: continue
    o = {a,P3[a],P5[a],P7[a]}; seen |= o
    orbsum.append(sum(lc[b] for b in o))
print("Galois orbits on the 48 lines:", len(orbsum), " (all free of size 4)")
print("rank Pic^G =", matrix(QQ, orbsum).rank())
E20 = identity_matrix(QQ,20)
def sig(a, v):
    "v in Pic-basis coords -> image under sigma_a"
    amb = vector(QQ,v)*Bm                       # ambient coords w.r.t. the 20 basis lines
    img = sum(amb[i]*vector(QQ,linecoord[perm(a)[bas[i]]]) for i in range(20))
    return vector(QQ, toPic(img))
A = {a: matrix(QQ,[sig(a,e) for e in E20.rows()]) for a in (3,5,7)}
A[1] = E20
print("each sigma_a acts by an isometry of Pic fixing H:",
      all(ip(vector(QQ,e)*A[a], vector(QQ,f)*A[a]) == ip(vector(QQ,e),vector(QQ,f))
          for a in (3,5,7) for e in E20.rows() for f in E20.rows())
      and all(vector(QQ,hG)*A[a] == vector(QQ,hG) for a in (3,5,7)))
```

Now the key observation. Suppose $\{E_1,\dots,E_{16}\}$ is a $G$-stable Kummer structure, $\Pi$ the
sublattice it spans. Then $\operatorname{Pic}_\mathbb{Q} = \Pi_\mathbb{Q} \oplus \Pi^\perp_\mathbb{Q}$
is a $G$-equivariant decomposition, so taking invariants is additive. The $E_i$ form an orthogonal basis
of $\Pi_\mathbb{Q}$ permuted by $G$, so $\dim \Pi^G_\mathbb{Q}$ is the number of $G$-orbits on them,
which is at least $16/|G| = 4$. And $H$ is $G$-invariant with nonzero projection $H_\perp$ to
$\Pi^\perp$ (indeed $H_\perp^2 = 4 + \frac12\sum d_i^2 > 0$), so $\dim(\Pi^\perp)^G_\mathbb{Q} \ge 1$.
Therefore
$$5 \;=\; \operatorname{rank}\operatorname{Pic}^G \;=\; \#\{\text{orbits}\} + \operatorname{rank}(\Pi^\perp)^G \;\ge\; 4 + 1 \;=\; 5,$$
and equality is forced throughout:

> **$G$ must act freely on the 16 curves**, in four orbits of size four, and $\operatorname{rank}(\Pi^\perp)^G = 1$.

No enumeration went into this; it holds in every degree. In particular each orbit consists of four
mutually disjoint curves of one and the same degree.

## 7. Isotypic reduction

A direct search over $(-2)$-classes is hopeless: enumerating those of degree $\le 4$ in the rank-20
lattice overflows a 24 GB PARI stack. But the freeness just proved makes the problem collapse.

Decompose $\operatorname{Pic}_\mathbb{Q}$ into isotypic pieces for $G$. Write $P_G = \frac14\sum_{g}g$
for the projection onto the invariants and $P_W = 1 - P_G$; distinct isotypic components are orthogonal.
For a curve $E$ in a free orbit put $E_1 = P_G(E)$ and $\xi = P_W(E)$. Then the orbit sum is
$$F \;=\; \textstyle\sum_g g(E) \;=\; 4E_1 ,$$
and orthogonality of $E$ to its three conjugates gives, since $E^2 = -2$ and $F^2 = -8$,
$$E_1^2 = -\tfrac12, \qquad \xi^2 = -\tfrac32, \qquad \xi\cdot g\xi = \tfrac12 \ \ (g \ne 1).$$
So $E$ is determined by a vector of norm $\tfrac32$ in the rank-15 negative definite lattice
$\Lambda_W = P_W(\operatorname{Pic})$ — a very short vector, hence few of them — together with a vector
of norm $\tfrac12$ in the rank-5 lattice $\Lambda_G = P_G(\operatorname{Pic})$, whose degree is
$E_1 \cdot H$. The rank-20 enumeration becomes a rank-5 one.

```python
import time
pari.allocatemem(8*10^9, silent=True)
Av = (A[1]+A[3]+A[5]+A[7])/4
BG = matrix(QQ, matrix(QQ,Av).row_module(ZZ).basis())
BW = matrix(QQ, matrix(QQ,E20-Av).row_module(ZZ).basis())
GrG = BG*GP*BG.transpose() ; GrW = BW*GP*BW.transpose()
print("rank Lambda_G =", BG.nrows(), ", rank Lambda_W =", BW.nrows(),
      ", Lambda_W negative definite:", (-GrW).is_positive_definite())

sW = lcm([QQ(x).denominator() for x in (-2*GrW).list()]+[1])
QW = matrix(ZZ, sW*(-2*GrW))
Xi = []
for r in pari(QW).qfminim(sW*3, 10^7, 2)[2].sage().transpose().rows():
    x = vector(ZZ,r)
    if x*QW*x == sW*3:
        for y in (x,-x): Xi.append(vector(QQ,y)*BW)
print("xi in Lambda_W with xi^2 = -3/2 :", len(Xi),
      ";  of these, xi.(g.xi) = 1/2 for every g != 1 :",
      sum(1 for x in Xi if all(ip(x, x*A[a]) == QQ(1)/2 for a in (3,5,7))))

cG  = vector(QQ,[ip(vector(QQ,BG.row(i)), vector(QQ,hG)) for i in range(5)])
QG0 = -2*GrG + matrix(QQ,[[cG[i]*cG[j] for j in range(5)] for i in range(5)])
sG  = lcm([QQ(x).denominator() for x in QG0.list()]+[1])
QG  = matrix(ZZ, sG*QG0)
print("the form -2v^2 + (v.H)^2 is positive definite on Lambda_G :", QG.is_positive_definite())
D = 4 ; t0 = time.time()
E1s = []
for r in pari(QG).qfminim(sG*(1+D*D), 10^7, 2)[2].sage().transpose().rows():
    for y in (vector(ZZ,r), -vector(ZZ,r)):
        v = vector(QQ,y)*BG
        if ip(v,v) == QQ(-1)/2 and 1 <= ip(v, vector(QQ,hG)) <= D: E1s.append(v)
print("E1 in Lambda_G with E1^2 = -1/2 and 1 <= deg <= %d :  %d   (%.0fs)" % (D, len(E1s), time.time()-t0))
```

## 8. The search

A curve of a free orbit is $E = E_1 + \xi$, subject to $E \in \operatorname{Pic}$ (integrality). Because
isotypic components are orthogonal and $E_1$ is $G$-invariant,
$$E \cdot g E' \;=\; E_1 \cdot E_1' \;+\; \xi \cdot g\xi' ,$$
so orthogonality of two whole orbits is four table look-ups. That makes the final search cheap.

```python
t0 = time.time()
curves = [(i,j) for i in range(len(E1s)) for j in range(len(Xi)) if all(c in ZZ for c in E1s[i]+Xi[j])]
byd = {}
for (i,j) in curves:
    d = ip(E1s[i], vector(QQ,hG)); byd[d] = byd.get(d,0)+1
print("(-2)-classes orthogonal to all their Galois conjugates, degree <= %d : %d" % (D,len(curves)))
print("   by degree:", dict(sorted(byd.items())))
print("   -> every such class has ODD degree; none is a conic.")

# one orbit per E1 (its four xi's form a single G-orbit)
from collections import defaultdict
byE1 = defaultdict(list)
for (i,j) in curves: byE1[i].append(j)
orb = sorted(byE1.items())
m = len(orb)
print("free G-orbits of size 4 :", m)

M1 = matrix(ZZ, [4*E1s[i] for (i,_) in orb])                    # 4*E1 = sum_g gE, integral
Q11 = M1*GP*M1.transpose()                                      # 16 * (E1_i . E1_j), in ZZ
MX  = matrix(ZZ, [4*x for x in Xi])                             # 4*xi, integral
PX  = {a: MX*GP*(MX*A[a]).transpose() for a in (1,3,5,7)}       # 16 * (xi_i . g xi_j)
# drop out of Sage matrices into plain Python lists: the adjacency scan below is
# ~10^6 iterations, and Sage's matrix __getitem__ dominates everything otherwise
Q11 = [[int(x) for x in r] for r in Q11.rows()]
PX  = {a: [[int(x) for x in r] for r in PX[a].rows()] for a in (1,3,5,7)}
rep = [byE1[i][0] for (i,_) in orb]
ONE = int(1)                                   # plain Python ints: Sage's preparser
adj = [int(0)]*m                               # would turn bit operations into Rationals
for i in range(m):
    ri = rep[i]; row = Q11[i]
    for j in range(i+1, m):
        v = row[j]; rj = rep[j]
        if all(v + PX[a][ri][rj] == 0 for a in (1,3,5,7)):
            adj[i] |= (ONE << int(j)); adj[j] |= (ONE << int(i))
def bits(x):
    x = int(x)
    while x:
        b = x & -x; yield int(b.bit_length()-1); x ^= b
def above(t): return ~((ONE << int(t+1)) - ONE)
npair = sum(bin(a).count('1') for a in adj)//2
print("orthogonal PAIRS of orbits:", npair, "  (%.0fs)" % (time.time()-t0))

tri = 0; quad = 0
for i in range(m):
    Ni = adj[i] & above(i)
    for j in bits(Ni):
        Nij = Ni & adj[j] & above(j)
        tri += bin(Nij).count('1')
        for k in bits(Nij):
            quad += bin(Nij & adj[k] & above(k)).count('1')
best = 4 if quad else (3 if tri else (2 if npair else 1))
print("orthogonal TRIPLES of orbits:", tri)
print()
print("="*72)
print("maximum number of mutually orthogonal free Galois orbits :", best)
print("   -> at most %d pairwise disjoint Galois-stable (-2)-classes, where 16 are needed" % (4*best))
print("4-tuples of mutually orthogonal free orbits (i.e. 16 disjoint classes) :", quad)
print("GALOIS-STABLE KUMMER STRUCTURES on the Fermat quartic, all degrees <= %d :  %d" % (D, quad))
print("="*72)
print("total %.0fs" % (time.time()-t0))
```

## 9. Does twisting help?

Every diagonal quartic $X_{a,b,c,d}: ax_0^4+bx_1^4+cx_2^4+dx_3^4 = 0$ is the Fermat quartic over
$\bar{\mathbb{Q}}$, via $\phi: x_i \mapsto a_i^{1/4}x_i$. What a twist changes is the *Galois action* on
$\operatorname{Pic}$: transporting along $\phi$ replaces $\sigma$ by $\delta_\sigma \circ \sigma$, where
$$\delta_\sigma \;=\; \bigl(\sigma(a_i^{1/4})/a_i^{1/4}\bigr)_i \;\in\; D = \mu_4^4/\mu_4 \cong (\mathbb{Z}/4)^3,$$
the diagonal automorphisms of the Fermat quartic. Since $\delta$ is a crossed homomorphism, each twist
gives a subgroup of $D \rtimes \operatorname{Gal}(\mathbb{Q}(\mu_8)/\mathbb{Q})$, of order 256. So the
obstruction of sections 5–8 might conceivably dissolve for a different twist.

The tractable family is the one where the lines stay defined over $\mathbb{Q}(\mu_8)$, so $|\Gamma|$ is
still 4 and the rank count of section 6 still applies. That means all ratios $a_i/a_j$ lie in
$\langle -1, 4\rangle \subset \mathbb{Q}^*/\mathbb{Q}^{*4}$ — a group of order 4, since $4^2 = 2^4$ — which
is exactly the subgroup generated by $-1$ and $4$ appearing in Ieronymou–Skorobogatov–Zarhin's $H_D$.
Normalising $a=1$ gives 64 twists.

```python
def p_sigma(mm): return [idx[(f,(mm*mu)%8,(mm*nu)%8)] for (f,mu,nu) in labels]
def p_delta(e):
    out=[]
    for (f,mu,nu) in labels:
        if   f=='L': u,v = 2*(e[0]-e[1]), 2*(e[2]-e[3])
        elif f=='M': u,v = 2*(e[0]-e[2]), 2*(e[1]-e[3])
        else:        u,v = 2*(e[0]-e[3]), 2*(e[1]-e[2])
        out.append(idx[(f,(mu+u)%8,(nu+v)%8)])
    return out
def compose(p,q): return [p[q[i]] for i in range(n)]
# e(a,m): exponent of zeta_4 in sigma_m(a^{1/4})/a^{1/4}
Etab = {1:{1:0,3:0,5:0,7:0}, -1:{1:0,3:1,5:2,7:3}, 4:{1:0,3:2,5:2,7:0}, -4:{1:0,3:3,5:0,7:3}}
def tperm(tw,mm): return compose(p_delta([Etab[tw[i]][mm] for i in range(4)]), p_sigma(mm))

import itertools
bad = sum(1 for tw in itertools.product([1,-1,4,-4],repeat=4) for m1 in (1,3,5,7) for m2 in (1,3,5,7)
          if compose(tperm(tw,m1),tperm(tw,m2)) != tperm(tw,(m1*m2)%8))
print("twisted action is a genuine group action (homomorphism failures):", bad)

def twrank(tw):
    P = {m: tperm(tw,m) for m in (1,3,5,7)}
    sums=[]; seen=set()
    for a in range(n):
        if a in seen: continue
        o = {P[m][a] for m in (1,3,5,7)}; seen |= o
        sums.append(sum(lc[b] for b in o))
    return matrix(QQ,sums).rank()
from collections import Counter
ranks = {tw: twrank(tw) for tw in itertools.product([1],[1,-1,4,-4],[1,-1,4,-4],[1,-1,4,-4])}
print("rank Pic^Gamma over the 64 twists:", dict(sorted(Counter(ranks.values()).items())))
print("   rank 4 => fewer than the 4 orbits required: %d twists die on the rank count alone"
      % sum(1 for r in ranks.values() if r == 4))
```

The 20 twists with $\operatorname{rank}\operatorname{Pic}^\Gamma = 4$ are ruled out immediately. Sweeping
the other 44 (degree $\le 6$ for the rank-5 cases, $\le 3$ for rank 8 and 9, where the invariant lattice
is large), 41 produce nothing at all. Three do produce configurations — and they are one surface up to
permuting coordinates:
$$x^4+y^4+4z^4+4w^4 = 0,$$
with $\operatorname{rank}\operatorname{Pic}^\Gamma = 9$, the largest in the family. It yields **256** sets
of 16 pairwise-orthogonal Galois-stable $(-2)$-classes with $2$-divisible sum, in six orbits of sizes
$2,2,2,2,4,4$, of degrees eight $2$'s and eight $3$'s.

Not all of these are configurations of *curves*, however: the search ranges over $(-2)$-classes, and an
effective class may be a reducible divisor. If $D$ is effective and $C$ is an irreducible curve with
$D \cdot C < 0$, then $C$ is a component of $D$; so an effective class $E$ of small degree with $E^2 = -2$
is reducible as soon as $E \cdot L < 0$ for some line $L$ — **excluding $L = E$ itself**, since every line
satisfies $E \cdot E = -2 < 0$ and is nevertheless perfectly irreducible. Testing that against the 48 lines:

```python
def shorts(Bl, targ):
    if targ == 0: return [vector(QQ,[0]*20)]
    Gr = Bl*GP*Bl.transpose(); sc = lcm([QQ(x).denominator() for x in (-2*Gr).list()]+[1])
    Q = matrix(ZZ, sc*(-2*Gr)); tg = sc*(-2*targ); out=[]
    for r in pari(Q).qfminim(tg, 10^7, 2)[2].sage().transpose().rows():
        x = vector(ZZ,r)
        if x*Q*x == tg:
            for y in (x,-x): out.append(vector(QQ,y)*Bl)
    return out
def act_matrix(perm):
    rows=[]
    for e in E20.rows():
        amb = vector(QQ,e)*Bm
        rows.append(vector(QQ, toPic(sum(amb[i]*vector(QQ,linecoord[perm[bas[i]]]) for i in range(20)))))
    return matrix(ZZ, rows)

tw = (1,1,4,4); Dt = 3
Mt = {m: act_matrix(tperm(tw,m)) for m in (3,5,7)}; Mt[1] = identity_matrix(ZZ,20)
Avt = (matrix(QQ,Mt[1])+Mt[3]+Mt[5]+Mt[7])/4
BGt = matrix(QQ, matrix(QQ,Avt).row_module(ZZ).basis())
BWt = matrix(QQ, matrix(QQ,E20-Avt).row_module(ZZ).basis()); rgt = BGt.nrows()
GrGt = BGt*GP*BGt.transpose()
cGt = vector(QQ,[ip(vector(QQ,BGt.row(i)), vector(QQ,hG)) for i in range(rgt)])
QG0 = -2*GrGt + matrix(QQ,[[cGt[i]*cGt[j] for j in range(rgt)] for i in range(rgt)])
sGt = lcm([QQ(x).denominator() for x in QG0.list()]+[1]); QGt = matrix(ZZ, sGt*QG0)
cand = set()
for st in (1,2,4):
    Xis = shorts(BWt, QQ(-2)+QQ(st)/2); e1t = QQ(-st)/2
    for r in pari(QGt).qfminim(sGt*(-2*e1t + Dt*Dt), 10^7, 2)[2].sage().transpose().rows():
        for y in (vector(ZZ,r), -vector(ZZ,r)):
            v = vector(QQ,y)*BGt
            if ip(v,v) != e1t or not (1 <= ip(v, vector(QQ,hG)) <= Dt): continue
            for x in Xis:
                Ec = v+x
                if all(c in ZZ for c in Ec): cand.add(tuple(ZZ(c) for c in Ec))
allC = [vector(ZZ,c) for c in cand]
irred = [Ec for Ec in allC if all(ip(Ec, lc[j]) >= 0
                                   for j in range(n) if tuple(lc[j]) != tuple(Ec))]
print("x^4+y^4+4z^4+4w^4 = 0 :  rank Pic^Gamma =", rgt)
print("   (-2)-classes of degree <= %d orthogonal to their conjugates :" % Dt, len(allC))
print("   of these, IRREDUCIBLE (no line component other than itself) :", len(irred))
print("   reducible, hence not curves                                 :", len(allC)-len(irred))
```

So 1424 of the 1664 candidates are reducible divisors, not curves, and 240 survive — the 48 lines among
them. Re-running the configuration search over those 240 classes — 92 Galois orbits — returns **16**, and
the same happens for the two coordinate permutations of this twist. Each of the 16 consists of eight lines
and eight conics. So this twist *does* carry a Galois-stable Kummer structure over $\mathbb{Q}$.

## 10. Conclusion

The search fails, and it fails earlier than the $2$-divisibility condition: one cannot even assemble 16
pairwise disjoint Galois-stable $(-2)$-classes. The maximum is **three** orbits — 12 classes, realised by
12 pairwise skew *lines* — where four orbits are required. It misses by exactly one orbit.

Two structural facts came out along the way:

- Every $(-2)$-class orthogonal to all of its Galois conjugates has **odd** degree. In particular no
  conic can belong to a Galois-stable configuration, so Swinnerton-Dyer's 8-lines-plus-8-conics shape is
  not merely unstable by accident — that shape can never be made stable.
- Any stable configuration would have to consist of four **free** Galois orbits (section 6), all four
  curves of an orbit sharing one degree.

Section 9 shows that twisting *does* rescue it: across the 64 diagonal twists that keep $|\Gamma| = 4$,
twenty die on the rank count, forty-one produce nothing, and three — one surface up to permuting
coordinates, $x^4+y^4+4z^4+4w^4 = 0$ — carry 16 genuine Galois-stable Kummer structures each.

So for the Fermat quartic itself no coboundary change of $\psi$ moves the cocycle into the affine
subgroup: $X$ is not $\operatorname{Kum}(V)$ over $\mathbb{Q}$ for any $2$-covering $V$ of $A$, and the
twisted Galois action of sections 1–2 cannot be transported to the abelian surface over $\mathbb{Q}$.
That is a statement about $\mathbb{Q}$, and about the Fermat quartic specifically: the twist
$x^4+y^4+4z^4+4w^4=0$ is $\operatorname{Kum}(V)$ over $\mathbb{Q}$, and the Fermat quartic itself is a
Kummer surface over $\mathbb{Q}(\sqrt2)$ and over $\mathbb{Q}(\sqrt{-2})$ — see the companion notebook
[fermat-quartic-kummer-fields.ipynb](fermat-quartic-kummer-fields.ipynb).

**Caveats, stated plainly.** In section 9 only the 64 twists with all $a_i/a_j \in \langle -1,4\rangle$
were examined; twists outside that group push the lines into a larger field, so $|\Gamma| > 4$ and the
rank inequality changes shape. The degree bounds there are $6$ (rank 5) and $3$ (rank 8, 9), and the
irreducibility test is exact only up to degree 3 — beyond that "no line component" is necessary but not
sufficient, though it only ever removes candidates, so the zeros are safe.

The enumeration for the Fermat quartic is complete only for degree $\le 12$; the rank-5 method
would push further, but this is not a proof for all degrees. Run externally at $D = 12$ the counts are
1404 free orbits, 7656 orthogonal pairs, maximum clique $3$, zero configurations. The cell above uses
$D = 4$ so that the notebook executes quickly; the conclusion is identical, and the degree-12 figures
quoted here come from the same code with the bound raised.

The other caveat is more interesting, and section 10 shows it has teeth. The search ranges over
$(-2)$-**classes**, not curves. For a negative answer that is conservative — every genuine configuration
of curves is among the classes tested, so finding none rules out more than was asked. For a *positive*
answer it is not enough: a class can be effective but reducible, and a reducible divisor is not one of
the 16 disjoint smooth rational curves a Kummer structure requires.
