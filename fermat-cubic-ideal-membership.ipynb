# The Fermat cubic surface and $A^2B^2C^2D^2 \in (A^3,B^3,C^3,D^3)$

Testing a proposed counterexample to a MathOverflow question of Hailong Dao
([Product vs sum in $\mathbb{C}[x,y,z]$](https://mathoverflow.net/questions/511761/product-vs-sum-in-mathbb-cx-y-z)):

> Let $R = \mathbb{C}[x,y,z]$. Prove or disprove: for all $A,B,C,D \in R$,
> $$A^2B^2C^2D^2 \in (A^3,B^3,C^3,D^3).$$

Briançon–Skoda gives the weaker $A^3B^3C^3D^3 \in (A^4,B^4,C^4,D^4)$; the question asks
to sharpen it by one.

The single answer to the question proposes a counterexample. Take the Fermat cubic surface
$X : F = X_1^3+X_2^3+X_3^3+X_4^3 = 0$ in $\mathbb{P}^3$. It is smooth, hence rational
(Clebsch), so there is a birational $\phi : \mathbb{P}^2 \dashrightarrow X$ given by four
forms $A,B,C,D$ of equal degree, which therefore satisfy
$$A^3+B^3+C^3+D^3 = 0 \quad\text{in } \mathbb{C}[x,y,z]. \tag{1}$$
Let $\varphi : S = \mathbb{C}[X_1,\dots,X_4] \to R$ send $X_i$ to $A,B,C,D$. Dominance of
$\phi$ gives $\ker\varphi = (F)$, and with $I = (X_1^3,\dots,X_4^3)$, $M = X_1^2X_2^2X_3^2X_4^2$
one has $M \notin I$ while $(F) \subseteq I$. The answer concludes
$\varphi(M) \notin \varphi(I) = (A^3,B^3,C^3,D^3)$.

**The flaw** (raised in the comments by Matthew Bolan): $\varphi(I)$ is the ideal generated by
$A^3,\dots,D^3$ inside the *subring* $\varphi(S) = \mathbb{C}[A,B,C,D]$, whereas
$(A^3,B^3,C^3,D^3)$ means the ideal in all of $R$ — its elements are
$f_1A^3+\dots+f_4D^3$ with $f_i \in \mathbb{C}[x,y,z]$ arbitrary, not necessarily in
$\operatorname{im}\varphi$. So the contradiction never gets off the ground.

This notebook does two things:

1. builds the construction explicitly and shows the membership **holds** for it, so the gap is
   not repairable;
2. gives a computation-free family satisfying every hypothesis of the argument for which the
   conclusion is provably false.

## 1. Lines on the Fermat cubic

The two-skew-lines construction: if $L_1, L_2 \subset X$ are skew, then for
$p \in L_1$, $q \in L_2$ the line $\overline{pq}$ meets the cubic in a third point, and
$(p,q) \mapsto r(p,q)$ is a birational map $\mathbb{P}^1\times\mathbb{P}^1 \dashrightarrow X$.
Writing $F(\alpha p + \beta q) = \alpha^2\beta\,\sigma_2 + \alpha\beta^2\sigma_1$ with
$\sigma_2 = 3\sum p_i^2q_i$, $\sigma_1 = 3\sum p_iq_i^2$ (the terms $\alpha^3F(p)$ and
$\beta^3F(q)$ vanish), the third root is $(\alpha:\beta) = (\sigma_1:-\sigma_2)$, so
$$r = \Big(\textstyle\sum p_iq_i^2\Big)\,p - \Big(\textstyle\sum p_i^2q_i\Big)\,q,$$
of bidegree $(2,2)$.

The 27 lines on $X$ are $x_i = -\omega^a x_j$, $x_k = -\omega^b x_l$ over the three pairings.
Only three of them are rational, and those pairwise **meet** — so the construction needs
$\omega$.

```python
K.<w> = CyclotomicField(3)          # w = primitive cube root of unity
S.<X1,X2,X3,X4> = K[]
F = X1^3 + X2^3 + X3^3 + X4^3

# the three rational lines, as spans of two points each
rational_lines = {
    "(12)(34)": (vector(K,[1,-1,0,0]), vector(K,[0,0,1,-1])),
    "(13)(24)": (vector(K,[1,0,-1,0]), vector(K,[0,1,0,-1])),
    "(14)(23)": (vector(K,[1,0,0,-1]), vector(K,[0,1,-1,0])),
}
names = list(rational_lines)
for i in range(3):
    for j in range(i+1, 3):
        p1,p2 = rational_lines[names[i]]
        q1,q2 = rational_lines[names[j]]
        print(names[i], "vs", names[j], "-> det =", matrix(K,[p1,p2,q1,q2]).det(), "(0 means they meet)")
```

So we twist the second line by $\omega$: take
$$L_1 = \{(s,-s,t,-t)\}, \qquad L_2 = \{(u,v,-\omega^2 u,-\omega v)\},$$
both on $X$, and now genuinely skew.

```python
p1, p2 = vector(K,[1,-1,0,0]), vector(K,[0,0,1,-1])       # L1
q1, q2 = vector(K,[1,0,-w^2,0]), vector(K,[0,1,0,-w])     # L2
print("skew? det =", matrix(K,[p1,p2,q1,q2]).det())

def third_point(P, Q):
    "third intersection of the line through P, Q with the Fermat cubic"
    s1 = sum(P[i]*Q[i]^2 for i in range(4))
    s2 = sum(P[i]^2*Q[i] for i in range(4))
    return [s1*P[i] - s2*Q[i] for i in range(4)]
```

## 2. The construction, in the answer's own setting

Compose $\mathbb{P}^1\times\mathbb{P}^1 \dashrightarrow X$ with the birational
$\mathbb{P}^2 \dashrightarrow \mathbb{P}^1\times\mathbb{P}^1$, $(x:y:z)\mapsto((x:z),(y:z))$.
The bidegree $(2,2)$ forms become quartic forms in $x,y,z$: a birational
$\phi : \mathbb{P}^2 \dashrightarrow X$ given by four forms of equal degree with no common
factor, exactly as the answer requires.

```python
R.<x,y,z> = K[]
P = [p1[i]*x + p2[i]*z for i in range(4)]     # L1 parametrized by (x:z)
Q = [q1[i]*y + q2[i]*z for i in range(4)]     # L2 parametrized by (y:z)
G = third_point(P, Q)
print("common factor:", gcd(G))
G = [p // gcd(G) for p in G]
A, B, C, D = G
for n, p in zip("ABCD", G):
    print(f"{n} (deg {p.degree()}) = {p}")
```

```python
# hypothesis (1): the Fermat relation
print("A^3+B^3+C^3+D^3 =", A^3 + B^3 + C^3 + D^3)

# hypothesis (2): ker(phi) = (F), i.e. phi is dominant onto X
phi = S.hom(G, R)
print("ker(phi) == (F)? ", phi.kernel() == S.ideal(F))
```

Both hypotheses of the answer hold. Now the actual question — membership in the ideal
of $R = \mathbb{C}[x,y,z]$, with arbitrary polynomial coefficients:

```python
I = R.ideal([A^3, B^3, C^3, D^3])
M = A^2*B^2*C^2*D^2
print("deg M =", M.degree(), "  deg A^3 =", (A^3).degree())
print("A^2B^2C^2D^2 in (A^3,B^3,C^3,D^3)? ", M in I)

cof = M.lift(I)                       # explicit certificate
print("cofactor degrees:", [f.degree() for f in cof])
print("certificate verifies:", sum(f*g for f, g in zip(cof, I.gens())) == M)
```

The membership **holds**, so the proposed counterexample is not one: the flagged gap is fatal
rather than cosmetic. Note also that the argument would work verbatim in the subring —
what it really proves is $A^2B^2C^2D^2 \notin (A^3,\dots,D^3)\,\mathbb{C}[A,B,C,D]$, a
restatement of $M \notin I$ that says nothing about $R$.

## 3. Not an artifact of one choice of lines

Same test over other skew pairs and other birational $\mathbb{P}^2\dashrightarrow\mathbb{P}^1\times\mathbb{P}^1$.

```python
def run(q1, q2, sub, label):
    p1, p2 = vector(K,[1,-1,0,0]), vector(K,[0,0,1,-1])
    if matrix(K,[p1,p2,q1,q2]).det() == 0:
        print(f"{label}: lines meet, skipped"); return
    (l1a, l1b), (l2a, l2b) = sub
    P = [p1[i]*l1a + p2[i]*l1b for i in range(4)]
    Q = [q1[i]*l2a + q2[i]*l2b for i in range(4)]
    G = third_point(P, Q)
    G = [p // gcd(G) for p in G]
    A, B, C, D = G
    assert A^3 + B^3 + C^3 + D^3 == 0
    dominant = S.hom(G, R).kernel() == S.ideal(F)
    member = A^2*B^2*C^2*D^2 in R.ideal([A^3, B^3, C^3, D^3])
    print(f"{label}:  degs={[p.degree() for p in G]}  ker=(F): {dominant}   membership: {member}")

run(vector(K,[1,0,-w^2,0]), vector(K,[0,1,0,-w]),   ((x,z),(y,z)),           "(13)(24) w-twist          ")
run(vector(K,[1,0,-w^2,0]), vector(K,[0,1,0,-w]),   ((x,y),(y,z)),           "same lines, other P^2 map ")
run(vector(K,[1,0,-w^2,0]), vector(K,[0,1,0,-w^2]), ((x,z),(y,z)),           "(13)(24) w^2-twist        ")
run(vector(K,[1,0,0,-w^2]), vector(K,[0,1,-w,0]),   ((x,z),(y,z)),           "(14)(23) twist            ")
run(vector(K,[1,-w,0,0]),   vector(K,[0,0,1,-w^2]), ((x,z),(y,z)),           "(12)(34) twist            ")
run(vector(K,[1,0,-w^2,0]), vector(K,[0,1,0,-w]),   ((x+z,z),(y+2*z,x-z)),   "generic substitution      ")
```

## 4. Why the argument cannot be valid, without any computation

The argument uses nothing about $A,B,C,D$ beyond (1) and $\ker\varphi = (F)$. So exhibit a
quadruple with both properties for which membership provably *holds*.

Let $A_0,\dots,D_0 \in \mathbb{C}[a,b]$ be an affine parametrization of $X$ (dehomogenize the
$(2,2)$ forms above), and set
$$A = c\,A_0(a,b),\quad \dots, \quad D = c\,D_0(a,b) \ \in\ \mathbb{C}[a,b,c].$$
Then $A^3+B^3+C^3+D^3 = c^3\cdot 0 = 0$, and the image of $\mathbb{C}^3 \to \mathbb{C}^4$ is
dense in the affine cone over $X$ (the projective image is dense in $X$, and $c$ sweeps the
scaling), so $\ker\varphi = (F)$ still.

But membership now reduces to a statement in **two** variables: since the generators lie in
$\mathbb{C}[a,b]$ and $c$ is free,
$$c^8 A_0^2B_0^2C_0^2D_0^2 \in c^3(A_0^3,\dots,D_0^3) \iff A_0^2B_0^2C_0^2D_0^2 \in (A_0^3,\dots,D_0^3) \text{ in } \mathbb{C}[a,b],$$
and that holds by Briançon–Skoda in dimension 2: with $J = (A_0^3,\dots,D_0^3)$ and
$m = \min_i v(A_{0,i})$, any valuation gives $v(A_0^2B_0^2C_0^2D_0^2) = 2\sum_i v(A_{0,i}) \ge 8m \ge 6m = v(J^2)$,
so the product lies in $\overline{J^2} \subseteq J$. (This is Dao's own comment on the question,
in the $n=2$ case.)

So the argument, if valid, would prove something false.

```python
R2.<a,b> = K[]
P0 = [p1[i]*a + p2[i] for i in range(4)]      # affine: t = 1
Q0 = [q1[i]*b + q2[i] for i in range(4)]      # affine: v = 1
G0 = third_point(P0, Q0)
A0, B0, C0, D0 = G0
print("A0^3+B0^3+C0^3+D0^3 =", A0^3 + B0^3 + C0^3 + D0^3)
print("[2 vars]  A0^2B0^2C0^2D0^2 in (A0^3,...,D0^3)? ",
      A0^2*B0^2*C0^2*D0^2 in R2.ideal([g^3 for g in G0]))
```

```python
# the scaled quadruple in three variables: all hypotheses of the answer, membership still holds
h = R2.hom([x, y], R)
G3 = [z*h(g) for g in G0]
A3_, B3_, C3_, D3_ = G3
print("identity:", A3_^3 + B3_^3 + C3_^3 + D3_^3 == 0)
print("ker(phi) == (F)? ", S.hom(G3, R).kernel() == S.ideal(F))
print("[3 vars]  A^2B^2C^2D^2 in (A^3,B^3,C^3,D^3)? ",
      A3_^2*B3_^2*C3_^2*D3_^2 in R.ideal([g^3 for g in G3]))
```

## Conclusion

Every hypothesis the answer invokes — the Fermat relation (1), dominance, $\ker\varphi = (F)$,
$M \notin I$ — is satisfied by explicit quadruples for which
$A^2B^2C^2D^2 \in (A^3,B^3,C^3,D^3)$ nonetheless. The step $\varphi(I) = (A^3,B^3,C^3,D^3)$
conflates the ideal in the subring $\mathbb{C}[A,B,C,D]$ with the ideal in
$\mathbb{C}[x,y,z]$, and the construction cannot be rescued.

The Fermat route is not a silly idea, though: because $D^3 = -(A^3+B^3+C^3)$, the relation
collapses $(A^3,B^3,C^3,D^3) = (A^3,B^3,C^3)$ — it *costs a generator*, which is what a
counterexample would want. It just happens that for these parametrizations the membership
survives anyway.

Dao's question remains open.
