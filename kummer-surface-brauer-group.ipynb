# The Brauer group of $x^4+4y^4=z^4+4w^4$

The surface
$$X:\quad x^4 + 4y^4 \;=\; z^4 + 4w^4 \qquad\text{i.e.}\qquad x^4 + 4y^4 - z^4 - 4w^4 = 0 \;\subset\; \mathbb{P}^3_{\mathbb{Q}}$$
is, after swapping $y$ and $z$, the surface $x^4 - y^4 + 4z^4 - 4w^4 = 0$ of section 19 of
[fermat-quartic-kummer-fields.ipynb](fermat-quartic-kummer-fields.ipynb): a diagonal quartic that is a Kummer
surface over $\mathbb{Q}$, namely $\operatorname{Kum}(T)$ for a nontrivial $2$-covering $T$ of an abelian surface
$A$ isogenous to $E\times E$, $E: y^2 = x^3 - x$
(see [kummer-abelian-surface-and-torsor.ipynb](kummer-abelian-surface-and-torsor.ipynb)). It has rational
points, e.g. $(1:0:1:0)$, and it contains the rational lines $x = \pm z,\ y = \pm w$.

Two questions:

1. Is $\operatorname{Br}(X)$ larger than the constant classes coming from $\operatorname{Br}(\mathbb{Q})$?
2. If so, does a Brauer class obstruct weak approximation?

**The answers.** $\operatorname{Br}(X) = \operatorname{Br}_0(X)$: there are no nonconstant classes at all, so the
second question has nothing to act on. The transcendental half of this is a theorem in the literature, which
section 2 quotes. The algebraic half is the group $H^1(\mathbb{Q}, \operatorname{Pic}\bar X)$, and computing it is
the actual content of this notebook: it is a finite computation with an integer lattice of rank $20$, done
twice below by two different routes and checked against Martin Bright's tables. The answer is $0$.

## 1. The filtration $\operatorname{Br}_0 \subseteq \operatorname{Br}_1 \subseteq \operatorname{Br}$

For a smooth projective variety $X$ over a number field $k$, with $\Gamma = \operatorname{Gal}(\bar k/k)$, put
$$\operatorname{Br}_0(X) = \operatorname{im}\bigl(\operatorname{Br}(k)\to\operatorname{Br}(X)\bigr), \qquad
  \operatorname{Br}_1(X) = \ker\bigl(\operatorname{Br}(X)\to\operatorname{Br}(\bar X)\bigr).$$
Classes in $\operatorname{Br}_1$ are **algebraic**: they die over $\bar k$. The rest are **transcendental**.
The two graded pieces are controlled as follows.

- **Algebraic.** The Hochschild–Serre spectral sequence for $\mathbb{G}_m$ on $X$, together with
  $H^0(\bar X, \mathbb{G}_m) = \bar k^*$ and $H^1(\bar X, \mathbb{G}_m) = \operatorname{Pic}\bar X$, gives an
  exact sequence
  $$\operatorname{Br}(k) \to \operatorname{Br}_1(X) \to H^1(k, \operatorname{Pic}\bar X) \to H^3(k, \mathbb{G}_m).$$
  For a number field $H^3(k,\mathbb{G}_m) = 0$, so
  $$\boxed{\ \operatorname{Br}_1(X)/\operatorname{Br}_0(X) \;\cong\; H^1(k, \operatorname{Pic}\bar X).\ }$$
- **Transcendental.** $\operatorname{Br}(X)/\operatorname{Br}_1(X)$ embeds in $\operatorname{Br}(\bar X)^\Gamma$.
  For a K3 surface $\operatorname{Br}(\bar X) \cong (\mathbb{Q}/\mathbb{Z})^{22-\rho}$, which here is
  $(\mathbb{Q}/\mathbb{Z})^2$ since $\rho(\bar X) = 20$.

**Why only $\operatorname{Br}/\operatorname{Br}_0$ matters for obstructions.** The Brauer–Manin pairing sends an
adelic point $(P_v)$ and a class $\alpha$ to $\sum_v \operatorname{inv}_v \alpha(P_v) \in \mathbb{Q}/\mathbb{Z}$.
A constant class $\alpha \in \operatorname{Br}_0$ pairs to $0$ with every adelic point, by the reciprocity law of
global class field theory. So $X(\mathbb{A}_k)^{\operatorname{Br}}$ depends only on
$\operatorname{Br}(X)/\operatorname{Br}_0(X)$, and if that group is $0$ then
$X(\mathbb{A}_k)^{\operatorname{Br}} = X(\mathbb{A}_k)$: there is no Brauer–Manin obstruction to anything.

## 2. The transcendental part: $\operatorname{Br}(X) = \operatorname{Br}_1(X)$

This part is quoted, not re-proved. Write the surface as $x_0^4 + a_1x_1^4 + a_2x_2^4 + a_3x_3^4 = 0$, so
$(a_1,a_2,a_3) = (4,-1,-4)$. Three results apply, and each gives the vanishing on its own.

- **Ieronymou–Skorobogatov–Zarhin, Corollary 3.3.** Let $H_D \subseteq \mathbb{Q}^*$ be generated by $-1$, $4$,
  $a_1,a_2,a_3$ and $\mathbb{Q}^{*4}$. If $\{2,3,5\}\cap H_D = \varnothing$ then $\operatorname{Br} = \operatorname{Br}_1$.
  For odd $\ell$ the proof runs through Mizukami's isomorphism of the Fermat quartic with a Kummer surface:
  $\operatorname{Br}(\bar X)$ becomes $\operatorname{Br}(\bar A)$, and the invariants are read off from the Galois
  action on the torsion of the CM curve $y^2 = x^3 - x$. The $2$-primary part ($2 \notin H_D$) is a theorem from
  Ieronymou's thesis.
- **Gvirtz–Skorobogatov, main theorem.** The $2$-primary part of $\operatorname{Br}(X)/\operatorname{Br}_1(X)$ is $0$
  unless $X$ is equivalent to $(1,2,-2)$ or $(1,8,-8)$, when it is $\mathbb{Z}/2$.
  Their Corollary also says directly: if $a_1a_2a_3$ is a square, $\operatorname{Br}(X) = \operatorname{Br}_1(X)$.
  Here $a_1a_2a_3 = 16$.
- **Ieronymou–Skorobogatov, Theorem 1.1.** The odd part is $0$ unless $-3a_1a_2a_3$ or $125a_1a_2a_3$ lies in
  $\mathbb{Q}^{*4} \cup -4\mathbb{Q}^{*4}$.

To decide "equivalent to $(1,2,-2)$" we only need an invariant: permuting, multiplying single coefficients by
fourth powers, and scaling all four by $\lambda$ change the product $a_0a_1a_2a_3$ by a fourth power. The two
exceptional surfaces have product class $-4$ in $\mathbb{Q}^*/\mathbb{Q}^{*4}$; ours has product class $16 \equiv 1$.

The cell checks the arithmetic hypotheses, using
$\mathbb{Q}^*/\mathbb{Q}^{*4} \cong \mathbb{Z}/2 \times \bigoplus_p \mathbb{Z}/4$ (the sign and the valuations mod $4$),
so that membership in a subgroup is a question about an integer lattice.

```python
def primes_of(x):
    x = QQ(x)
    return set(ZZ(x.numerator()).prime_divisors()) | set(ZZ(x.denominator()).prime_divisors())

def mod4_vector(x, primes):
    # Q^*/Q^*4 = {+-1} x (+)_p Z/4 : record the sign and the valuations
    x = QQ(x)
    return [0 if x > 0 else 1] + [x.valuation(p) for p in primes]

def in_class_group(target, gens):
    """Is target in the subgroup of Q^*/Q^*4 generated by gens?"""
    primes = sorted(set().union(primes_of(target), *[primes_of(g) for g in gens]) | {2})
    rel = diagonal_matrix(ZZ, [2] + [4]*len(primes))
    L = matrix(ZZ, [mod4_vector(g, primes) for g in gens] + [[0]*(1+len(primes))]).stack(rel).row_module()
    return vector(ZZ, mod4_vector(target, primes)) in L

def fourth_power_class(x): return in_class_group(x, [])

a = (4, -1, -4)                       # x^4 + 4y^4 - z^4 - 4w^4, normalised so a_0 = 1
H_D = [-1, 4] + list(a)
print("ISZ Cor. 3.3 :  2, 3, 5 in H_D ?", [in_class_group(q, H_D) for q in (2, 3, 5)])
P = prod(a)
print("GS, 2-primary: a1a2a3 = %s ; same class as -4 (the class of (1,2,-2) and (1,8,-8)) ?" % P,
      fourth_power_class(P/(-4)))
for c in (-3, 125):
    print("IS, odd part : %4d*a1a2a3 in Q*^4 or -4Q*^4 ?" % c,
          fourth_power_class(c*P) or fourth_power_class(c*P/(-4)))
```

All hypotheses hold, so $\operatorname{Br}(X) = \operatorname{Br}_1(X)$, and
$$\operatorname{Br}(X)/\operatorname{Br}_0(X) \;\cong\; H^1(\mathbb{Q}, \operatorname{Pic}\bar X).$$
The rest of the notebook computes the right-hand side.

## 3. Reducing $H^1(\mathbb{Q},\operatorname{Pic}\bar X)$ to a finite group

$H^1(\mathbb{Q}, \operatorname{Pic}\bar X)$ is a continuous cohomology group of the profinite group
$\Gamma_{\mathbb{Q}}$. Two observations make it finite and explicit.

**The splitting field is $k = \mathbb{Q}(\mu_8)$.** Over $\bar{\mathbb{Q}}$, the map
$\phi_u: x_i \mapsto u_ix_i$ with $u_i^4 = a_i$ identifies $X$ with the Fermat quartic
$x_0^4 + x_1^4 + x_2^4 + x_3^4 = 0$. Here all four fourth roots can be taken in $k$, with $\epsilon = \zeta_8$:
$$1^{1/4} = 1,\qquad 4^{1/4} = \sqrt2 = \epsilon + \epsilon^{-1},\qquad (-1)^{1/4} = \epsilon,\qquad (-4)^{1/4} = 1 + i .$$
The $48$ lines of the Fermat quartic are defined over $k$ and generate $\operatorname{Pic}$ (Pinch–Swinnerton-Dyer).
Since $\phi_u$ is defined over $k$, the $48$ lines of $X$ are defined over $k$ too. So $\Gamma_k$ acts trivially on
$\operatorname{Pic}\bar X$.

**Inflation–restriction.** With $G = \operatorname{Gal}(k/\mathbb{Q}) = \{\sigma_1,\sigma_3,\sigma_5,\sigma_7\} \cong (\mathbb{Z}/2)^2$,
where $\sigma_m(\epsilon) = \epsilon^m$,
$$0 \to H^1(G, \operatorname{Pic}\bar X) \to H^1(\mathbb{Q}, \operatorname{Pic}\bar X) \to H^1(k, \operatorname{Pic}\bar X)
   = \operatorname{Hom}_{\mathrm{cts}}(\Gamma_k, \mathbb{Z}^{20}) = 0,$$
the last group being $0$ because a continuous homomorphism from a profinite group to $\mathbb{Z}$ has finite image.
So $H^1(\mathbb{Q},\operatorname{Pic}\bar X) = H^1(G, \operatorname{Pic}\bar X)$: the first cohomology of a group of
order $4$ acting on $\mathbb{Z}^{20}$.

The same argument shows that only the image of $G$ in $\operatorname{GL}_{20}(\mathbb{Z})$ matters (inflate from the
quotient by the kernel, a finite group, which has no nonzero homomorphisms to $\mathbb{Z}^{20}$). This justifies
treating $G$ below simply as a set of four integer matrices.

## 4. The Picard lattice

Each line is the row space of a $2\times 4$ matrix over $k$. For $\mu,\nu$ odd,
$$L_{\mu\nu}: x_0 = \epsilon^\mu x_1,\ x_2 = \epsilon^\nu x_3, \qquad
  M_{\mu\nu}: x_0 = \epsilon^\mu x_2,\ x_1 = \epsilon^\nu x_3, \qquad
  N_{\mu\nu}: x_0 = \epsilon^\mu x_3,\ x_1 = \epsilon^\nu x_2 .$$
Two distinct lines meet (intersection number $1$) exactly when their two bases together span less than $k^4$;
every line has self-intersection $-2$. Choose $20$ lines spanning $\operatorname{Pic}\otimes\mathbb{Q}$, express
every line in terms of them, and take the $\mathbb{Z}$-span of all $48$: that is $\operatorname{Pic}\bar X$, with a
$\mathbb{Z}$-basis `Bm` and Gram matrix `GP`. The discriminant should be $-64$.

```python
K.<ep> = CyclotomicField(8)              # ep = zeta_8
odds = [1, 3, 5, 7]
spec = {'L': (0,1,2,3), 'M': (0,2,1,3), 'N': (0,3,1,2)}
lines, labels = [], []
for fam in ['L', 'M', 'N']:
    c = spec[fam]
    for mu in odds:
        for nu in odds:
            B = matrix(K, 2, 4)
            B[0,c[0]] = ep^mu; B[0,c[1]] = 1      # x_{c0} = ep^mu x_{c1}
            B[1,c[2]] = ep^nu; B[1,c[3]] = 1      # x_{c2} = ep^nu x_{c3}
            lines.append(B); labels.append((fam, mu, nu))
n = len(lines)
R.<s,t> = PolynomialRing(K)
print("%d lines, all on the Fermat quartic:" % n,
      all(sum((s*B[0,j] + t*B[1,j])^4 for j in range(4)) == 0 for B in lines))

Gram = matrix(ZZ, n, n, lambda i, j: -2 if i == j else (1 if lines[i].stack(lines[j]).rank() < 4 else 0))
bas = []
for i in range(n):
    if Gram[bas+[i], bas+[i]].rank() == len(bas) + 1: bas.append(i)
GB = Gram[bas, bas]
coord = [GB.inverse()*vector(QQ, [Gram[i,j] for j in bas]) for i in range(n)]
Bm = matrix(QQ, matrix(QQ, coord).row_module(ZZ).basis())     # Z-basis of Pic, in line coordinates
GP = Bm*GB*Bm.transpose()
print("Pic: rank %d, discriminant %d" % (GP.nrows(), GP.det()))
```

## 5. The Galois action, read straight off the lines

Transport the Galois action of $X$ to the Fermat quartic along $\phi_u$. A line $L$ on the Fermat quartic is sent to
$$\sigma \cdot L \;=\; \phi_u\bigl(\sigma(\phi_u^{-1}(L))\bigr):$$
pull $L$ back to $X$ (divide the $j$-th coordinate by $u_j$), apply $\sigma$ to the coefficients, and push forward
again (multiply by $u_j$). The result is another of the $48$ lines, recognised by its reduced row echelon form.

This computes the permutation from the geometry, so it does not depend on the sign conventions of the closed
formula $\sigma \mapsto \delta_\sigma^{-1}\circ\sigma$ used in the companion notebooks.

A permutation of the lines extends linearly to $\operatorname{Pic}$. **Convention:** the $i$-th row of `M[m]` is the
image of the $i$-th basis vector, so $\sigma_m$ acts on row vectors by $x \mapsto xM_m$. The cell checks that each
`M[m]` is an isometry and that $m \mapsto M_m$ is a faithful action of the Klein four-group.

```python
u = {1: K(1), -1: ep, 4: ep + ep^-1, -4: 1 + ep^2}
assert all(r^4 == c for c, r in u.items())
sigma = {m: K.hom([ep^m]) for m in (1, 3, 5, 7)}
where = {tuple(B.echelon_form().list()): i for i, B in enumerate(lines)}

def galois_perm(coeffs, m):
    us = [u[c] for c in coeffs]
    perm = []
    for B in lines:
        C = matrix(K, 2, 4, lambda r, j: us[j]*sigma[m](B[r,j]/us[j]))
        perm.append(where[tuple(C.echelon_form().list())])
    return perm

def act_matrix(perm):
    rows = []
    for e in identity_matrix(QQ, 20).rows():
        amb = e*Bm
        rows.append(Bm.solve_left(sum(amb[i]*coord[perm[bas[i]]] for i in range(20))))
    return matrix(ZZ, rows)

def galois_matrices(coeffs):
    return {m: act_matrix(galois_perm(coeffs, m)) for m in (1, 3, 5, 7)}

A = (1, 4, -1, -4)
M = galois_matrices(A)
I20 = identity_matrix(ZZ, 20)
print("permutations of the 48 lines   :", all(sorted(galois_perm(A, m)) == list(range(n)) for m in (3,5,7)))
print("isometries of Pic              :", all(M[m]*GP*M[m].transpose() == GP for m in (3,5,7)))
print("sigma_1 acts trivially         :", M[1] == I20)
print("a Klein four-group             :", M[3]*M[5] == M[7] and all(M[m]^2 == I20 for m in (3,5,7)))
print("faithful                       :", len({tuple(M[m].list()) for m in (1,3,5,7)}) == 4)

def invariant_rank(Ms):
    # x is invariant iff x*(g - 1) = 0 for every g: put the conditions side by side
    return block_matrix(QQ, [[g - 1 for g in Ms]], subdivide=False).left_kernel().dimension()
print("rank Pic^Gamma                 :", invariant_rank(M.values()))
```

$\operatorname{rank}\operatorname{Pic}(\bar X)^G = 9$: that is the rank of $\operatorname{Pic}(X)$ over $\mathbb{Q}$,
and it agrees with the value in section 19 of the companion notebook, which was computed from the closed formula.

## 6. $H^1$ from the definition

Let $G$ be a finite group acting on $M = \mathbb{Z}^r$ (on column vectors, so we use the transposes of the matrices
above). Then
$$Z^1 = \{ f: G \to M \;:\; f(gh) = f(g) + g\,f(h) \ \ \forall g,h \}, \qquad
  B^1 = \{ g \mapsto gm - m \;:\; m \in M \}, \qquad H^1 = Z^1/B^1 .$$
Everything here is integer linear algebra.

- A $1$-cochain is a vector $(f(g_1),\dots,f(g_N)) \in \mathbb{Z}^{Nr}$, and the cocycle condition is a system of
  $N^2 r$ linear equations. $Z^1$ is the integer kernel. A kernel inside $\mathbb{Z}^{Nr}$ is automatically
  saturated, so `right_kernel_matrix` gives a genuine $\mathbb{Z}$-basis.
- $B^1$ is spanned by the coboundaries of the standard basis vectors $e_1,\dots,e_r$.
- Writing those coboundaries in the basis of $Z^1$ gives an integer matrix whose Smith normal form is the
  structure of $Z^1/B^1$.

Two sanity checks are built in. First, $H^1(G,M)$ is killed by $|G|$, hence finite, so $B^1$ must have full rank in
$Z^1$. Second, over $\mathbb{Q}$ the group $H^1$ vanishes, so $\operatorname{rank} Z^1 = \operatorname{rank} B^1
= \operatorname{rank}\bigl(M/M^G\bigr) = r - \operatorname{rank} M^G$.

```python
def H1(group):
    """H^1(G, Z^r) for a finite group G given as a list of r x r integer matrices, acting on columns.
    Returns (rank of Z^1, invariant factors of Z^1/B^1)."""
    G = [matrix(ZZ, g) for g in group]
    key = lambda g: tuple(g.list())
    index = {key(g): i for i, g in enumerate(G)}
    assert len(index) == len(G) and all(key(g*h) in index for g in G for h in G), "not a group"
    r = G[0].nrows(); I = identity_matrix(ZZ, r)
    # a 1-cochain is (f(g_0), ..., f(g_{N-1})) in Z^{N r}; cocycle: f(gh) - f(g) - g f(h) = 0
    conditions = []
    for g in G:
        for h in G:
            blocks = [zero_matrix(ZZ, r) for _ in G]
            blocks[index[key(g*h)]] += I
            blocks[index[key(g)]]   -= I
            blocks[index[key(h)]]   -= g
            conditions.append(block_matrix(ZZ, [blocks], subdivide=False))
    Z1 = block_matrix(ZZ, [[c] for c in conditions], subdivide=False).right_kernel_matrix()
    # coboundaries f(g) = g m - m, for m running over the standard basis
    B1 = block_matrix(ZZ, [[g - I] for g in G], subdivide=False).transpose()
    X = Z1.solve_left(B1)
    assert X.denominator() == 1
    d = [e for e in matrix(ZZ, X).elementary_divisors() if e != 0]
    assert len(d) == Z1.nrows(), "Z^1/B^1 is not finite"
    return Z1.nrows(), [e for e in d if e != 1]

def show(inv): return "0" if not inv else " x ".join("Z/%d" % e for e in inv)

G = [M[m].transpose() for m in (1, 3, 5, 7)]
rZ, inv = H1(G)
print("rank Z^1 = %d  (= 20 - rank Pic^Gamma = %d)" % (rZ, 20 - invariant_rank(M.values())))
print("H^1(Q, Pic Xbar) =", show(inv))
```

## 7. A second route: the presentation of $(\mathbb{Z}/2)^2$

The same group, computed differently. $G = \langle s, t \mid s^2 = t^2 = 1,\ st = ts\rangle$ with $s = \sigma_3$,
$t = \sigma_5$. A cocycle is determined by $a = f(s)$ and $b = f(t)$, and the three relations translate into
$$f(s^2) = a + sa = 0, \qquad f(t^2) = b + tb = 0, \qquad f(st) = f(ts):\ a + sb = b + ta .$$
So
$$Z^1 = \{(a,b) \in M^2 : (1+s)a = 0,\ (1+t)b = 0,\ (1-t)a = (1-s)b\}, \qquad
  B^1 = \{((s-1)m,\ (t-1)m) : m \in M\}.$$
Now the system has $40$ unknowns instead of $80$. It shares no code with the previous cell beyond the matrices
themselves.

```python
S, T = M[3].transpose(), M[5].transpose()
O = zero_matrix(ZZ, 20)
C = block_matrix(ZZ, [[I20 + S, O], [O, I20 + T], [I20 - T, S - I20]], subdivide=False)
Z1 = C.right_kernel_matrix()
B1 = block_matrix(ZZ, [[S - I20], [T - I20]], subdivide=False).transpose()   # rows: ((s-1)m, (t-1)m)
X = Z1.solve_left(B1)
d = [e for e in matrix(ZZ, X).elementary_divisors() if e != 0]
assert len(d) == Z1.nrows()
print("from the presentation <s, t | s^2, t^2, st = ts> :", show([e for e in d if e != 1]))
```

## 8. Checking against Bright's tables

Bright's thesis classifies the Galois action on $\operatorname{Pic}\bar V$ for every diagonal quartic over $\mathbb{Q}$
into $546$ cases, and records $\rho = \operatorname{rank} H^0(k,\operatorname{Pic}\bar V)$ and the structure of
$\operatorname{Br}_1 V/\operatorname{Br}_0 V \cong H^1(k,\operatorname{Pic}\bar V)$. Table A.1 is the case
$k \cap \mathbb{Q}(\epsilon) = \mathbb{Q}$. The relevant rows:

| Bright | coefficients | $\rho$ | $\operatorname{Br}_1/\operatorname{Br}_0$ | surface |
|---|---|---|---|---|
| A62 | $[1,1,1,1]$ | 5 | $(\mathbb{Z}/2)^4$ | the Fermat quartic |
| A64 | $[1,1,4,4]$ | 9 | $0$ | the Kummer surface of section 11 of the companion notebook |
| A74 | $[1,1,-1,-1]$ | 9 | $\mathbb{Z}/2$ | $x^4+y^4 = z^4+w^4$ |
| **A75** | $[1,-4,4,-1]$ | **9** | $\mathbf{0}$ | **ours**, up to permuting coordinates |

Running both our coefficient vector and Bright's representative for A75 checks that the computation is
insensitive to permuting coordinates. The three other rows check that the code can see nonzero classes:
a program that always printed $0$ would fail on A62 and A74.

```python
bright = {(1,4,-1,-4): ("ours", 9, []), (1,-4,4,-1): ("A75", 9, []), (1,1,1,1): ("A62", 5, [2,2,2,2]),
          (1,1,4,4): ("A64", 9, []), (1,1,-1,-1): ("A74", 9, [2])}
print("%-16s %-6s %-12s %-12s" % ("coefficients", "Bright", "rank Pic^G", "H^1"))
for coeffs, (case, rho, br) in bright.items():
    Mc = galois_matrices(coeffs)
    rk = invariant_rank(Mc.values()); _, inv = H1([Mc[m].transpose() for m in (1,3,5,7)])
    ok = (rk == rho and inv == br)
    print("%-16s %-6s %-12s %-12s %s" % (coeffs, case, rk, show(inv), "matches" if ok else "MISMATCH"))
```

All five rows match Bright exactly, in both columns.

## 9. Over the quadratic subfields

The computation also gives the algebraic Brauer group over each quadratic subfield $k' \subset \mathbb{Q}(\mu_8)$.
Over $k'$ the relevant group is $\operatorname{Gal}(k/k') = \langle\sigma_m\rangle$, and the argument of section 3
applies unchanged. For comparison, the same is done for $x^4+y^4 = z^4+w^4$.

```python
names = {3: "Q(sqrt(-2))", 5: "Q(i)", 7: "Q(sqrt(2))"}
for coeffs in [(1,4,-1,-4), (1,1,-1,-1)]:
    Mc = galois_matrices(coeffs)
    for m in (3, 5, 7):
        rk = invariant_rank([Mc[m]]); _, inv = H1([I20, Mc[m].transpose()])
        print("%-16s over %-12s rank Pic^G = %2d, H^1 = %s" % (coeffs, names[m], rk, show(inv)))
```

Three observations.

**Over $\mathbb{Q}(i)$ nothing survives.** That is the field where the $2$-covering $T$ becomes trivial, so there
$X \cong \operatorname{Kum}(A)$ for an honest abelian surface (section 6 of the torsor notebook).

**Over $\mathbb{Q}(\sqrt2)$ and $\mathbb{Q}(\sqrt{-2})$ there is a class of order $2$, and it does not descend to
$\mathbb{Q}$.** The reason is visible in the Hochschild–Serre sequence for $H = \langle\sigma_m\rangle \subset G$:
$$0 \to H^1(G/H, M^H) \to H^1(G, M) \to H^1(H, M)^{G/H} \xrightarrow{\ d_2\ } H^2(G/H, M^H).$$
Here $H^1(H,M) = \mathbb{Z}/2$, which is automatically $G/H$-invariant because $\mathbb{Z}/2$ has no nontrivial
automorphisms. Since $H^1(G,M) = 0$, the transgression $d_2$ is injective: the class over $k'$ is Galois-invariant,
but there is an obstruction in $H^2(G/H, M^H)$ to lifting it to a class over $\mathbb{Q}$.

**These are statements about $\operatorname{Br}_1$ only.** The theorems quoted in section 2 concern $\mathbb{Q}$
(and, in Gvirtz–Skorobogatov, $\mathbb{Q}(i)$). Nothing here says anything about transcendental classes over
$\mathbb{Q}(\sqrt{\pm2})$.

Note also that over $\mathbb{Q}$ the two surfaces differ, $0$ against $\mathbb{Z}/2$, while over every quadratic
subfield they give the same answer.

## 10. Conclusion

Putting sections 2 and 6–8 together,
$$\boxed{\ \operatorname{Br}(X) \;=\; \operatorname{Br}_0(X) \;\cong\; \operatorname{Br}(\mathbb{Q})\ }
\qquad\text{for}\quad X:\ x^4+4y^4 = z^4+4w^4 .$$
The map $\operatorname{Br}(\mathbb{Q}) \to \operatorname{Br}(X)$ is injective because $X$ has a rational point. By
section 1, $X(\mathbb{A}_\mathbb{Q})^{\operatorname{Br}} = X(\mathbb{A}_\mathbb{Q})$: **there is no Brauer–Manin
obstruction to weak approximation on $X$.**

That is not a proof that weak approximation holds. For K3 surfaces it is conjectured (Skorobogatov) that the
Brauer–Manin obstruction is the only one, which would give $X(\mathbb{Q})$ dense in $X(\mathbb{A}_\mathbb{Q})$. I
know of no unconditional proof for this surface. Ieronymou–Skorobogatov–Zarhin raise the same question for the
family $x^4 + 4y^4 + cz^4 - cw^4 = 0$, which also has $\operatorname{Br} = \operatorname{Br}(\mathbb{Q})$.

The Kummer structure does not change this. The class $\tau \in H^1(\mathbb{Q}, A[2])$ of the $2$-covering is
nontrivial, but no Brauer class on $X$ comes out of it.

### What is computed, and what is quoted

- **Computed here:** $H^1(\mathbb{Q}, \operatorname{Pic}\bar X) = 0$. The Galois action is derived directly from the
  lines, the answer is obtained by two independent linear-algebra routes, and the method reproduces four entries of
  Bright's table, including two nonzero ones.
- **Quoted:** $\operatorname{Br}(X) = \operatorname{Br}_1(X)$ (ISZ Corollary 3.3, independently Gvirtz–Skorobogatov
  and Ieronymou–Skorobogatov). The cell in section 2 checks only that their hypotheses hold.
- **Also used:** that the $48$ lines generate $\operatorname{Pic}$ of the Fermat quartic (Pinch–Swinnerton-Dyer),
  confirmed here by the discriminant $-64$.
- **Not settled:** whether weak approximation actually holds on $X$.

A natural next case: $x^4 + y^4 = z^4 + w^4$ has $\operatorname{Br}_1/\operatorname{Br}_0 = \mathbb{Z}/2$
(Bright A74, reproduced above). Whether that class obstructs weak approximation is a genuine computation: write
down an Azumaya algebra and evaluate its local invariants.
