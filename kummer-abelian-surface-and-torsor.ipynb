# The abelian surface behind a diagonal quartic

The companion notebook [fermat-quartic-kummer-fields.ipynb](fermat-quartic-kummer-fields.ipynb) shows that
$$X:\; x^4 - y^4 + 4z^4 - 4w^4 = 0$$
is a Kummer surface over $\mathbb{Q}$ — it carries $16$ pairwise disjoint irreducible $(-2)$-curves with
$2$-divisible sum, permuted by $\operatorname{Gal}(\mathbb{Q}(\mu_8)/\mathbb{Q})$ — and that Galois fixes
none of those $16$ curves, so what sits underneath is not an abelian surface but a **torsor**.

This notebook asks what exactly that torsor and its abelian surface are. The answer turns out to be
completely readable off the Picard lattice, and section 1 explains why before any computation happens.

Everything here is self-contained: the $48$ lines, the lattice, the Galois action and the Kummer structures
are all rebuilt from scratch.

## 1. What the Picard lattice remembers

Throughout, $F$ is a field of characteristic $0$ with $\Gamma = \operatorname{Gal}(\bar F/F)$.

### 1.1 Kummer surfaces of 2-coverings

Let $A$ be an abelian surface over $F$. A **2-covering** of $A$ is a torsor $T$ under $A$ whose class in
$H^1(F,A)$ comes from a class $\tau \in H^1(F,A[2])$; concretely $T$ carries a morphism $T \to A$ which over
$\bar F$ is a twist of multiplication by $2$. On such a $T$ the involution $\iota$ "$v \mapsto -v$" is
defined over $F$ even though there is no origin, and

$$\operatorname{Kum}(T) \;=\; \widetilde{T/\iota}$$

is a K3 surface over $F$. Blowing down its $16$ exceptional curves recovers $T$. Write
$$\Omega \;=\; \{E_1,\dots,E_{16}\}$$
for those curves. They are the exceptional curves over the fixed points of $\iota$, and the fixed-point set
$T^{\iota}$ is precisely the fibre of $T \to A$ over $0$. So:

> **$\Omega$ is a torsor under $A[2]$, and its class in $H^1(F,A[2])$ is the class $\tau$ of the 2-covering.**

In particular $\Omega$ has a $\Gamma$-fixed point exactly when $\tau = 0$, i.e. exactly when
$\operatorname{Kum}(T) \cong \operatorname{Kum}(A)$ for an honest abelian surface. That is the criterion used
throughout the companion notebook.

Conversely — this is Nikulin's theorem — any K3 surface over $\bar F$ with $16$ disjoint smooth rational
curves whose sum is $2$-divisible is such a $\operatorname{Kum}$, and if the set of $16$ is $\Gamma$-stable
the whole picture descends to $F$.

### 1.2 Reading off $A$: product, isogenous to a product, or neither

Inside $\operatorname{Pic}(\bar Y)$, $Y = \operatorname{Kum}(T)$, the $16$ curves span a sublattice of rank
$16$; let $N$ be its primitive closure (the *Nikulin lattice*) and
$$M \;=\; N^{\perp} \subset \operatorname{Pic}(\bar Y),$$
of rank $20-16 = 4$. The blow-down identifies

**Proposition 1.** $M \cong \operatorname{NS}(\bar A)(2)$ as lattices *and* as $\Gamma$-modules: the
intersection form on $M$ is twice the one on $\operatorname{NS}(\bar A)$, and the Galois action on $M$ is
the Galois action on $\operatorname{NS}(\bar A)$.

So $\operatorname{NS}(\bar A)$ with its Galois action is available for free once the $16$ curves are known.
(Since $\operatorname{NS}(\bar A)$ is even, $M$ has all its values divisible by $4$; and
$\det M = 2^4 \det\operatorname{NS}(\bar A)$.) Note $\operatorname{NS}(\bar T) = \operatorname{NS}(\bar A)$,
so it makes no difference that we started from a torsor.

Now recall that for an abelian surface $A$ over $F$ we have $\operatorname{NS}(A) = \operatorname{NS}(\bar A)^{\Gamma}$
— the usual obstruction lies in $\operatorname{Br}(F)$ and vanishes because $A$ has the rational point $0$.
So the Galois-invariant part of $M$, halved, **is** the Néron–Severi group of $A$ over $F$, and:

**Proposition 2.** Let $A$ be an abelian surface over $F$.
1. $A$ is isogenous over $F$ to a product of elliptic curves $\iff$
   $\operatorname{NS}(\bar A)^{\Gamma}$ contains a nonzero isotropic class.
2. $A \cong E \times E'$ over $F$ for elliptic curves $E,E'$ $\iff$
   $\operatorname{NS}(\bar A)^{\Gamma}$ contains a **hyperbolic plane**: isotropic $f_1,f_2$ with
   $f_1\cdot f_2 = 1$.

*Proof.* (1, $\Leftarrow$) Let $f \ne 0$ with $f^2 = 0$. Riemann–Roch on an abelian surface reads
$\chi(f) = f^2/2 = 0$ and $h^2(f) = h^0(-f)$, so one of $\pm f$ is effective; fix the sign with
$f\cdot h > 0$ for an invariant ample $h$, and replace $f$ by a primitive multiple. Then $|f|$ is a pencil
without fixed part and $f^2 = 0$, so it defines a fibration $A \to \mathbb{P}^1$ over $F$ whose fibres are
genus-1 curves. The fibre through $0$ is an elliptic curve $E \subset A$ defined over $F$, and
$E' = A/E$ is another, with $A$ isogenous to $E\times E'$. ($\Rightarrow$) Pull back a fibre class along the
isogeny.

(2, $\Leftarrow$) Let $f_1,f_2$ be as stated, giving elliptic curves $E_1,E_2 \subset A$ through $0$ with
$E_1\cdot E_2 = f_1\cdot f_2 = 1$. The addition map $E_1\times E_2 \to A$ is an isogeny of degree
$E_1 \cdot E_2 = 1$, hence an isomorphism over $F$. ($\Rightarrow$) The two fibre classes of a product meet
once. $\;\blacksquare$

Three regimes are therefore possible for one and the same $\bar A$: a hyperbolic plane (product), isotropic
but no hyperbolic plane (isogenous to a product only), and anisotropic (neither). The third really occurs;
deciding between them is a question about whether a ternary quadratic form over $\mathbb{Q}$ represents
zero, which Hasse–Minkowski settles and PARI's `qfsolve` answers directly.

**A warning.** $\operatorname{NS}(\bar A)^{\Gamma}$ is an indefinite lattice of small rank whose Gram matrix
can have large entries, so searching a box of small vectors is *not* a reliable way to look for isotropic
classes. Use `qfsolve`.

### 1.3 Reading off the torsor: the $16$ curves as an affine space

The class $\tau \in H^1(F,A[2])$ is likewise visible in the lattice. The key is that the affine structure on
$\Omega$ is not extra data — it is determined by which subsets of the $16$ curves have $2$-divisible sum.

**Proposition 3.** For $S \subseteq \Omega$ write $e_S = \sum_{E \in S} E$. Then
$$\mathcal{C} \;=\; \bigl\{\, S \subseteq \Omega \;:\; \tfrac12 e_S \in \operatorname{Pic}(\bar Y) \,\bigr\}$$
is an $\mathbb{F}_2$-linear code of length $16$ and dimension $5$, with weight enumerator
$1 + 30\,x^8 + x^{16}$: it is the first-order Reed–Muller code $\mathrm{RM}(1,4)$. Its $30$ words of weight
$8$ are exactly the $30$ affine hyperplanes of an affine $\mathbb{F}_2$-structure on $\Omega$, under which
$\Omega \cong \mathbb{A}^4(\mathbb{F}_2)$ compatibly with the $A[2]$-action.

*Why.* Identify $\Omega$ with $A[2] \cong \mathbb{F}_2^4$ over $\bar F$. Nikulin's computation of the
Kummer lattice says the $2$-divisible subsets are $\varnothing$, all of $\Omega$, and the $30$ affine
hyperplanes — that is precisely $\mathrm{RM}(1,4)$, whose words are the affine-linear functions
$\mathbb{F}_2^4 \to \mathbb{F}_2$. Being defined by a $2$-divisibility condition in $\operatorname{Pic}$,
$\mathcal{C}$ is visible on the lattice alone, and the affine structure can be read back off it.

Note that $\mathcal{C}$ is computed by linear algebra over $\mathbb{F}_2$, not by trying $2^{16}$ subsets:
$S \mapsto e_S \bmod 2$ is $\mathbb{F}_2$-linear, and $\mathcal{C}$ is its kernel.

**Proposition 4.** $\Gamma$ permutes $\Omega$ and preserves $\mathcal{C}$, hence acts on $\Omega$ by affine
transformations. Choosing $\omega_0 \in \Omega$ as origin identifies $\Omega$ with $A[2]$ and gives
$$\sigma(\omega) \;=\; \rho(\sigma)\,\omega \;+\; c(\sigma), \qquad \rho(\sigma) \in \mathrm{GL}_4(\mathbb{F}_2),\;\; c(\sigma) \in A[2],$$
where $\rho$ is the Galois representation on $A[2]$ (independent of $\omega_0$) and $c$ is a $1$-cocycle,
$c(\sigma\tau) = c(\sigma) + \rho(\sigma)c(\tau)$. Changing $\omega_0$ changes $c$ by a coboundary, so
$$[c] \in H^1(F, A[2])$$
is well defined, and by section 1.1 it **is** the class $\tau$ of the 2-covering $T$.

**Proposition 5.** For any extension $F'/F$: $[c]|_{\Gamma_{F'}} = 0 \iff \Omega$ has a $\Gamma_{F'}$-fixed
point $\iff Y_{F'} \cong \operatorname{Kum}(A')$ for an honest abelian surface $A'$ over $F'$. So the
smallest fields over which the torsor dies are read off directly from the orbit structure of $\Omega$.

**Proposition 6 (a consistency check).** If $A \cong E\times E'$ over $F$ then $A[2] = E[2]\oplus E'[2]$, so
$\rho$ must preserve a $2$-dimensional subspace. A $\rho$ with no stable plane rules out a product.

One caveat worth stating now: a K3 surface can carry several different Kummer structures, and when it does,
the different $\Omega$'s can give **different** $F$-forms $A$ and different classes $\tau$ — even though
over $\bar F$ the abelian surface is the same. So "the" abelian surface behind $X$ is not well posed until a
Kummer structure is fixed. Section 4 shows this happening.

## 2. The surface, the lines, and the Picard lattice

Over $\bar{\mathbb{Q}}$, $X: x^4-y^4+4z^4-4w^4=0$ becomes the Fermat quartic $x_0^4+x_1^4+x_2^4+x_3^4=0$
via $x_i \mapsto u_i x_i$ with $u_i^4 = (1,-1,4,-4)_i$. So we build the Fermat quartic's lattice, and carry
the $\mathbb{Q}$-structure of $X$ across as a twisted Galois action (section 3).

The Fermat quartic contains $48$ lines, and Pinch and Swinnerton-Dyer proved they generate
$\operatorname{Pic}$, which has rank $20$ and discriminant $-64$.

```python
K.<ep> = CyclotomicField(8)              # ep = zeta_8
odds = [1,3,5,7]
spec = {'L': (0,1,2,3), 'M': (0,2,1,3), 'N': (0,3,1,2)}
lines = []
for fam in ['L','M','N']:
    c = spec[fam]
    for mu in odds:
        for nu in odds:
            B = matrix(K,2,4); B[0,c[0]] = ep^mu; B[0,c[1]] = 1; B[1,c[2]] = ep^nu; B[1,c[3]] = 1
            lines.append((fam,mu,nu,B))
n = len(lines)
R.<s,t> = PolynomialRing(K)
bad = sum(1 for (f,mu,nu,B) in lines
          if sum((s*B[0][j] + t*B[1][j])^4 for j in range(4)) != 0)
print("lines: %d,  not lying on the Fermat quartic: %d" % (n, bad))
Gram = matrix(ZZ,n,n)
for a in range(n):
    for b in range(n):
        Gram[a,b] = -2 if a==b else (1 if lines[a][3].stack(lines[b][3]).rank() < 4 else 0)
labels = [(f,mu,nu) for (f,mu,nu,B) in lines]
idx = {labels[a]:a for a in range(n)}
print("rank of the lattice they span:", Gram.rank())
```

```python
bas = []
for a in range(n):
    if Gram[bas+[a],bas+[a]].rank() == len(bas)+1: bas.append(a)
GB = Gram[bas,bas]; GBinv = GB.inverse()
linecoord = [GBinv*vector(QQ,[Gram[a,b] for b in bas]) for a in range(n)]
PicQ = matrix(QQ, linecoord).row_module(ZZ)
Bm = matrix(QQ, PicQ.basis())                  # a Z-basis of Pic
GP = Bm*GB*Bm.transpose()
def toPic(v): return Bm.solve_left(matrix(QQ,[v]))[0]
def ip(u,v): return u*GP*v
lc = [vector(ZZ, toPic(linecoord[a])) for a in range(n)]
PicZ = matrix(QQ, [list(toPic(linecoord[a])) for a in range(n)]).row_module(ZZ)
hG = vector(QQ, toPic(sum(linecoord[idx[('L',1,nu)]] for nu in odds)))     # hyperplane class
E20 = identity_matrix(QQ,20)
print("Pic: rank %d, discriminant %s" % (GP.nrows(), GP.det()))
print("H^2 = %s, and H meets every line in %s point" % (ip(hG,hG), {ip(hG,L) for L in lc}))
```

### Irreducible curve classes in low degree

A Kummer structure consists of *curves*, not merely classes, so we need to recognise irreducibility. Since
$E^2 = -2$ makes $E$ rigid, the effective divisor in $|E|$ is unique, and $E$ is irreducible exactly when it
is not a sum of two nonzero effective classes. Building complete lists in degrees $1,2,3$ makes that test
exact up to degree $4$, which is what the search below needs.

```python
w  = GP*hG
Q0 = -2*GP + matrix(QQ,[list(w)]).transpose()*matrix(QQ,[list(w)])   # positive definite on Pic
print("Q = -2*(intersection form) + (.H)^2 is positive definite :", Q0.is_positive_definite())
sc = lcm([QQ(x).denominator() for x in Q0.list()]+[1]); Qi = matrix(ZZ, sc*Q0)
def classes(d, sq):
    out = []
    for r in pari(Qi).qfminim(sc*(-2*sq + d^2), 10^7, 2)[2].sage().transpose().rows():
        for y in (vector(ZZ,r), -vector(ZZ,r)):
            if ip(y,y) == sq and ip(y,hG) == d: out.append(y)
    return out
EFF1 = {tuple(x) for x in lc}
IRR2 = [F for F in classes(2,-2) if not any(tuple(F-L) in EFF1 for L in lc)]
EFF2 = {tuple(vector(ZZ,a)+vector(ZZ,b)) for a in EFF1 for b in EFF1} | {tuple(C) for C in IRR2}
IRR3 = [F for F in classes(3,-2)+classes(3,0)
        if not (any(tuple(F-L) in EFF2 for L in lc) or any(tuple(F-C) in EFF1 for C in IRR2))]
EFF3 = {tuple(vector(ZZ,a)+vector(ZZ,b)) for a in EFF1 for b in EFF2} | {tuple(C) for C in IRR3}
print("irreducible: %d lines, %d conics, %d cubics" % (len(lc), len(IRR2), len(IRR3)))
def eff_rr(B): return ip(B,hG) > 0 and ip(B,B) >= -2        # Riemann-Roch certificate
def irreducible(E):
    d = ip(E,hG)
    if d <= 1: return True
    for A_ in lc:
        B = E - A_
        if ip(B,hG) > 0 and (tuple(B) in EFF1 or tuple(B) in EFF2 or tuple(B) in EFF3 or eff_rr(B)):
            return False
    if d >= 4:
        for A_ in IRR2:
            B = E - A_
            if ip(B,hG) > 0 and (tuple(B) in EFF1 or tuple(B) in EFF2 or eff_rr(B)): return False
        for A_ in IRR3:
            B = E - A_
            if ip(B,hG) > 0 and (tuple(B) in EFF1 or eff_rr(B)): return False
    return True
```

## 3. The twisted Galois action, and the $16$ Kummer structures

Carrying the $\mathbb{Q}$-structure of $X_{1,-1,4,-4}$ across $\phi: x_i \mapsto u_ix_i$ replaces the
Galois action $\sigma$ on the Fermat quartic by $\delta_\sigma^{-1}\circ\sigma$, where
$\delta_\sigma = (\sigma(u_i)/u_i)_i$ lies in the group $D = \mu_4^4/\mu_4$ of diagonal automorphisms. Both
$\sigma$ and $D$ permute the $48$ lines, so the whole action is a permutation of $48$ objects.

```python
def p_sigma(m): return [idx[(f,(m*mu)%8,(m*nu)%8)] for (f,mu,nu) in labels]
def p_delta(e):
    out = []
    for (f,mu,nu) in labels:
        if   f=='L': a_,b_ = 2*(e[0]-e[1]), 2*(e[2]-e[3])
        elif f=='M': a_,b_ = 2*(e[0]-e[2]), 2*(e[1]-e[3])
        else:        a_,b_ = 2*(e[0]-e[3]), 2*(e[1]-e[2])
        out.append(idx[(f,(mu+a_)%8,(nu+b_)%8)])
    return out
def compose(p,q): return [p[q[i]] for i in range(n)]
def act_matrix(perm):
    rows=[]
    for e in E20.rows():
        amb = vector(QQ,e)*Bm
        rows.append(vector(QQ, toPic(sum(amb[i]*linecoord[perm[bas[i]]] for i in range(20)))))
    return matrix(ZZ, rows)
# sigma_m(u)/u for the chosen fourth roots of 1, -1, 4, -4
Etab = {}
for v,u in {1:K(1), -1:ep, 4:ep+ep^-1, -4:1+ep^2}.items():
    assert u^4 == v
    Etab[v] = {1:0}
    for m in (3,5,7):
        r = K.hom([ep^m])(u)/u
        Etab[v][m] = [j for j in range(4) if r == ep^(2*j)][0]
A = (1,-1,4,-4)
def tperm(m): return compose(p_delta([Etab[A[i]][m] for i in range(4)]), p_sigma(m))
M = {m: act_matrix(tperm(m)) for m in (3,5,7)}; M[1] = identity_matrix(ZZ,20)
print("Galois group law  sigma_3 sigma_5 = sigma_7 :", M[3]*M[5] == M[7])
print("each is an involution                       :", all(M[m]*M[m]==M[1] for m in (3,5,7)))
print("rank Pic^Gamma =",
      matrix(QQ, sum([list((M[m]-M[1]).rows()) for m in (3,5,7)],[])).right_kernel().dimension())
```

```python
import time
def shorts(Bl, targ):
    if targ == 0: return [tuple(vector(QQ,[0]*20))]
    Gr = Bl*GP*Bl.transpose(); s2 = lcm([QQ(x).denominator() for x in (-2*Gr).list()]+[1])
    Q = matrix(ZZ, s2*(-2*Gr)); tg = s2*(-2*targ); out=[]
    for r in pari(Q).qfminim(tg, 10^7, 2)[2].sage().transpose().rows():
        x = vector(ZZ,r)
        if x*Q*x == tg:
            for y in (x,-x): out.append(tuple(vector(QQ,y)*Bl))
    return out
t0 = time.time(); D = 4
Av = (matrix(QQ,M[1])+M[3]+M[5]+M[7])/4
BG = matrix(QQ, matrix(QQ,Av).row_module(ZZ).basis())
BW = matrix(QQ, matrix(QQ,E20-Av).row_module(ZZ).basis()); rg = BG.nrows()
GrG = BG*GP*BG.transpose()
cG  = vector(QQ,[ip(vector(QQ,BG.row(i)), hG) for i in range(rg)])
QG0 = -2*GrG + matrix(QQ,[[cG[i]*cG[j] for j in range(rg)] for i in range(rg)])
sG  = lcm([QQ(x).denominator() for x in QG0.list()]+[1]); QG = matrix(ZZ, sG*QG0)
cand = set()
for ss in (1,2,4):
    e1t = QQ(-ss)/2; Xis = shorts(BW, QQ(-2)+QQ(ss)/2)
    for r in pari(QG).qfminim(sG*(-2*e1t + D*D), 10^7, 2)[2].sage().transpose().rows():
        y = vector(ZZ,r)
        if y*GrG*y != e1t: continue
        dg = y*cG
        if   1 <= dg <= D: pass
        elif 1 <= -dg <= D: y = -y
        else: continue
        v = vector(QQ,y)*BG
        for x in Xis:
            E = tuple(v[i]+x[i] for i in range(20))
            if all(c_ in ZZ for c_ in E): cand.add(tuple(ZZ(c_) for c_ in E))
C = [E for E in (vector(ZZ,c_) for c_ in cand) if irreducible(E)]
key = {tuple(c_):i for i,c_ in enumerate(C)}; seen=set(); orbs=[]
for i,c_ in enumerate(C):
    if i in seen: continue
    img = [key.get(tuple(c_*M[m])) for m in (1,3,5,7)]
    if any(t_ is None for t_ in img): continue
    o = sorted(set(img)); seen |= set(o)
    if all(ip(C[a_],C[b_])==0 for a_ in o for b_ in o if a_<b_): orbs.append(o)
m_ = len(orbs)
Rm = matrix(ZZ,[C[o[0]] for o in orbs]); PR = {q: Rm*GP*(Rm*M[q]).transpose() for q in (1,3,5,7)}
comp = [[all(PR[q][i,j]==0 for q in (1,3,5,7)) for j in range(m_)] for i in range(m_)]
sols=[]
def rec(st,ch,tot):
    if tot==16:
        S=sum(C[t_] for o in ch for t_ in orbs[o])
        if all(x%2==0 for x in S): sols.append(list(ch))
        return
    if tot>16 or len(ch)>=rg: return
    for i in range(st,m_):
        if len(orbs[i])+tot>16: continue
        if all(comp[i][j] for j in ch):
            ch.append(i); rec(i+1,ch,tot+len(orbs[i])); ch.pop()
rec(0,[],0)
STRUCTS = [[C[t_] for o in ch for t_ in orbs[o]] for ch in sols]
print("Galois-stable Kummer structures of degree <= %d : %d   [%.0fs]" % (D, len(STRUCTS), time.time()-t0))
V0 = STRUCTS[0]
print("   one of them: degrees %s" % sorted(ip(E,hG) for E in V0))
print("   all (-2) %s ; disjoint %s ; 2-divisible %s ; all irreducible %s"
      % (all(ip(E,E)==-2 for E in V0),
         all(ip(V0[i],V0[j])==0 for i in range(16) for j in range(i+1,16)),
         (sum(V0)/2) in PicZ, all(irreducible(E) for E in V0)))
```

## 4. The abelian surface $A$

Now apply Propositions 1 and 2. For each of the $16$ structures, take the orthogonal complement of the $16$
curves, halve the form to get $\operatorname{NS}(\bar A)$, restrict to the Galois invariants, and ask
`qfsolve` whether the resulting ternary form represents zero.

```python
Bp = matrix(QQ, PicZ.basis())
def NS_invariants(Vc):
    Msys = matrix(QQ, [[ip(vector(QQ,Bp.row(i)), E) for E in Vc] for i in range(20)])
    ker  = Msys.left_kernel().basis_matrix()*Bp
    L    = matrix(QQ, (matrix(QQ,ker).row_module(ZZ) & PicZ).basis())
    NS   = (L*GP*L.transpose())/2                      # Proposition 1
    acts = [matrix(QQ,[vector(QQ, L.solve_left(matrix(QQ,[vector(QQ,L.row(i))*M[m]]))[0])
                       for i in range(4)]) for m in (3,5,7)]
    I4 = identity_matrix(QQ,4)
    inv = matrix(QQ, sum([list((a-I4).rows()) for a in acts],[])).right_kernel()
    B = matrix(QQ, (matrix(QQ, inv.basis()).row_module(ZZ) & (ZZ^4)).basis())
    return L, NS, B, B*NS*B.transpose()
L0, NS0, B0, Q0inv = NS_invariants(V0)
print("complement of the 16 curves: rank %d, det %s, all entries even %s"
      % (L0.nrows(), (L0*GP*L0.transpose()).det(), all(x%2==0 for x in (L0*GP*L0.transpose()).list())))
print("so det NS(Abar) = %s  -- Shioda-Mitani's value for E_i x E_2i" % NS0.det())
print("rank NS(Abar)^Gamma = %d, Gram" % B0.nrows()); print(Q0inv)
```

```python
from collections import Counter
def classify(Vc):
    L,NS,B,Q = NS_invariants(Vc); r = B.nrows()
    d = lcm([QQ(x).denominator() for x in Q.list()]+[1]); Qi_ = matrix(ZZ, d*Q)
    try:
        x = vector(ZZ, pari(Qi_).qfsolve().sage()); isot = (x != 0 and x*Qi_*x == 0)
    except (TypeError, ValueError):
        isot = False; x = None
    hyp = False
    if isot:
        g = gcd(list(x)); u = vector(ZZ,[c_//g for c_ in x])
        hyp = (gcd([ZZ(u*Qi_*vector(ZZ,e)) for e in identity_matrix(ZZ,r).rows()]) == d)
    return r, isot, hyp
tal = Counter(classify(Vc) for Vc in STRUCTS)
print("over the %d Kummer structures:" % len(STRUCTS))
for (r,isot,hyp),k in sorted(tal.items()):
    what = ("A is a PRODUCT E x E' over Q" if hyp else
            ("A is isogenous to a product, but not a product" if isot else
             "A is not even isogenous to a product over Q"))
    print("   %2d structures : rank NS^Gamma %d, isotropic %-5s, hyperbolic plane %-5s -> %s"
          % (k, r, isot, hyp, what))
```

So the answer depends on the structure — exactly the phenomenon flagged at the end of section 1. Over
$\bar{\mathbb{Q}}$ every one of them gives the same $\bar A \cong E_i \times E_{2i}$ (that is forced: $X$ is
a singular K3, so Shioda–Inose determines it from its transcendental lattice, and Shioda–Mitani determines
$\bar A$ from that). What varies is the $\mathbb{Q}$-form.

When $A$ *is* a product, the two factors are constrained: $E$ must be a **quartic** twist of $y^2=x^3-x$
(which has $j=1728$, so $\operatorname{Aut} = \mu_4$), while $E'$ can only be a **quadratic** twist of the
$j=287496$ curve, since $j \ne 0,1728$ leaves $\operatorname{Aut} = \{\pm1\}$. Pinning down *which* twists
needs more than the lattice — it needs the transcendental side — and is not attempted here.

## 5. The torsor $T$

Now Propositions 3 and 4. First the code $\mathcal{C}$, computed as an $\mathbb{F}_2$-kernel.

```python
F2 = GF(2)
Mred = matrix(F2, [[ZZ(c_) % 2 for c_ in E] for E in V0])      # 16 x 20
code_ = Mred.left_kernel()
print("dim C = %d over F_2" % code_.dimension())
wts = Counter(sum(int(c_) for c_ in wd) for wd in code_)
print("weight distribution %s   -- RM(1,4) is 1 + 30x^8 + x^16" % dict(sorted(wts.items())))
words = [tuple(int(c_) for c_ in wd) for wd in code_]
hyps  = [wd for wd in words if sum(wd) == 8]
print("affine hyperplanes: %d" % len(hyps))
o0 = 0                                                  # choose a curve as origin
through = [wd for wd in hyps if wd[o0] == 1]
print("hyperplanes through the chosen origin: %d (expect 15)" % len(through))
basisH = []
for wd in through:
    if matrix(F2,[list(x) for x in basisH+[wd]]).rank() == len(basisH)+1: basisH.append(wd)
    if len(basisH) == 4: break
coords = {j: vector(F2,[1-basisH[i][j] for i in range(4)]) for j in range(16)}
print("phi : Omega -> F_2^4 is a bijection :", len({tuple(v) for v in coords.values()}) == 16)
print("phi(origin) =", tuple(coords[o0]))
```

```python
import itertools
pos = {tuple(E): j for j,E in enumerate(V0)}
rho = {}; cc = {}
for m in (3,5,7):
    perm = [pos[tuple(E*M[m])] for E in V0]
    c_ = coords[perm[o0]]
    rows = []
    for i in range(4):
        e = vector(F2,[1 if k==i else 0 for k in range(4)])
        j = [k for k in range(16) if coords[k] == e][0]
        rows.append(coords[perm[j]] - c_)
    Rr = matrix(F2, rows).transpose()
    rho[m] = Rr; cc[m] = c_
    print("sigma_%d :  c = %s   rho =" % (m, tuple(c_)))
    print(Rr)
    print("            affine on all 16 points :", all(coords[perm[j]] == Rr*coords[j] + c_ for j in range(16)))
print()
print("cocycle identity c(st) = c(s) + rho(s)c(t) :",
      cc[7] == cc[3] + rho[3]*cc[5] and rho[7] == rho[3]*rho[5])
I4f = identity_matrix(F2,4)
triv = any(all(cc[m] == (rho[m]-I4f)*vector(F2,b) for m in (3,5,7))
           for b in itertools.product([0,1],repeat=4))
print("the class [c] in H^1(Gal, A[2]) is trivial :", triv)
print("rho-stable 2-dimensional subspaces of A[2] :",
      len([W for W in VectorSpace(F2,4).subspaces(2)
           if all(rho[m]*ww in W for m in (3,5,7) for ww in W.basis())]),
      " (a product needs at least one -- Proposition 6)")
```

## 6. Where the torsor dies

By Proposition 5 the subfields over which $X$ becomes the Kummer surface of an honest abelian surface are
read off from the fixed points of the subgroups. The cocycle above already shows what to expect:
$c(\sigma_5) = 0$, so $\sigma_5$ fixes the chosen origin.

```python
field = {3:"Q(sqrt-2)", 5:"Q(i)", 7:"Q(sqrt2), and R"}
for m in (3,5,7):
    fx = [j for j,E in enumerate(V0) if tuple(E*M[m]) == tuple(E)]
    print("sigma_%d  (fixed field %-16s) : %2d of the 16 curves fixed  ->  %s"
          % (m, field[m], len(fx),
             "Kum(ABELIAN SURFACE) there" if fx else "Kum(torsor) only"))
allfx = [j for j,E in enumerate(V0) if all(tuple(E*M[m])==tuple(E) for m in (3,5,7))]
print("full Galois group                          : %2d fixed  ->  %s"
      % (len(allfx), "Kum(abelian surface) over Q" if allfx else "Kum(torsor) over Q"))
print()
print("orbit sizes of the full group on the 16 curves:")
grp = [M[1],M[3],M[5],M[7]]; seen=set(); sz=[]
for E in V0:
    if tuple(E) in seen: continue
    o = {tuple(E*g) for g in grp}; seen |= o; sz.append(len(o))
rk = matrix(QQ, sum([list((M[m]-M[1]).rows()) for m in (3,5,7)],[])).right_kernel().dimension()
print("   %s -> %d orbits ; rank identity %d + %d = %d" % (sorted(sz), len(sz), len(sz), rk-len(sz), rk))
```

## 7. Summary

For $X: x^4-y^4+4z^4-4w^4 = 0$ over $\mathbb{Q}$, with the $16$ Kummer structures of degree $\le 4$:

**The abelian surface.** Over $\bar{\mathbb{Q}}$ it is $E_i \times E_{2i}$, forced by Shioda–Inose and
Shioda–Mitani. Over $\mathbb{Q}$ the form depends on the structure chosen: $7$ of the $16$ give an honest
product $E\times E'$, $4$ give something isogenous to a product but not a product, and $5$ give something
not even isogenous to a product. In the product cases $E$ is a quartic twist of $y^2=x^3-x$ and $E'$ a
quadratic twist of the $j = 287496$ curve.

**The torsor.** The $16$ curves carry a canonical affine $\mathbb{F}_2$-structure, recovered from the
$2$-divisible subsets, which form $\mathrm{RM}(1,4)$ exactly as Proposition 3 predicts. Galois acts
affinely, and the resulting cocycle $c$ represents the class of the $2$-covering in $H^1(\mathbb{Q},A[2])$.
It is nontrivial, so $X$ is not $\operatorname{Kum}$ of an abelian surface over $\mathbb{Q}$.

**Where it dies.** $c(\sigma_5) = 0$: over $\mathbb{Q}(i)$ the torsor becomes trivial and $X$ *is* the
Kummer surface of an honest abelian surface. Over $\mathbb{Q}(\sqrt2)$, $\mathbb{Q}(\sqrt{-2})$,
$\mathbb{R}$ and $\mathbb{Q}$ it remains a nontrivial torsor.

**Caveats.** The $16$ structures are those with curves of degree $\le 4$; the irreducibility test is exact
in that range and not beyond, so a structure using higher-degree curves would be invisible here and could
in principle give a different answer. The torsor cocycle is computed for one chosen structure; the others
will generally give different classes. And identifying the specific twists $E$, $E'$ is left open — the
lattice does not see it.
