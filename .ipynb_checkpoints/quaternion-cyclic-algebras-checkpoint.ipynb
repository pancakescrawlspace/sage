# Quaternion algebras and cyclic algebras in Sage

A worked tour of central simple algebra computations in SageMath 10.9.

Sage ships a full-featured `QuaternionAlgebra` (degree 2), but **no** general cyclic
algebra class. So the plan is: exercise the built-in quaternion machinery first, then
build cyclic algebras $(K/F,\sigma,b)$ of arbitrary degree by hand and check the
construction against Sage's quaternions in the degree-2 case.

**Contents**

1. Hamilton's quaternions
2. General $(a,b)$ and where they ramify
3. Splitting: zero divisors and reduction mod $p$
4. Orders, units, and class numbers
5. Quaternion algebras over number fields
6. Cyclic algebras of arbitrary degree
7. Coda: odd valuations and ramification

## 1. Hamilton's quaternions

$B = \left(\frac{-1,-1}{\mathbb{Q}}\right) = \mathbb{Q}\langle i,j\rangle$ with
$i^2 = -1$, $j^2 = -1$, $ji = -ij$, and $k := ij$.

```python
B.<i,j,k> = QuaternionAlgebra(-1, -1)
print(B)
print("basis:", B.basis())
print("i^2 =", i^2, " j^2 =", j^2, " k^2 =", k^2)
print("ij =", i*j, "  ji =", j*i, "  jk =", j*k, "  ki =", k*i)

gens = [B(1), i, j, k]
mult_table = matrix(B, 4, 4, [[u*v for v in gens] for u in gens])
print("\nmultiplication table (rows/cols in order 1, i, j, k):")
show(mult_table)
```

The **reduced norm** and **reduced trace** are $\mathrm{Nrd}(z) = z\bar z$ and
$\mathrm{Trd}(z) = z + \bar z$; every $z$ satisfies its reduced characteristic
polynomial $X^2 - \mathrm{Trd}(z)X + \mathrm{Nrd}(z)$.

```python
z = 1 + 2*i - 3*j + k
print("z      =", z)
print("zbar   =", z.conjugate())
print("Nrd(z) =", z.reduced_norm(), "   (= z*zbar =", z*z.conjugate(), ")")
print("Trd(z) =", z.reduced_trace())

f = z.reduced_characteristic_polynomial()
print("reduced charpoly:", f)
print("f(z) = 0 ?", f(z) == 0)
print("z^-1   =", z.inverse(), "  check:", z*z.inverse() == 1)
```

Left multiplication gives the **regular representation** $B \hookrightarrow M_4(\mathbb{Q})$.
Its determinant is $\mathrm{Nrd}(z)^2$ and its trace is $2\,\mathrm{Trd}(z)$ — the regular
representation is the reduced one, doubled.

```python
M = z.matrix()
show(M)
print("det  =", M.det(), " = Nrd^2 =", z.reduced_norm()^2)
print("trace =", M.trace(), " = 2*Trd =", 2*z.reduced_trace())
```

Multiplicativity of the reduced norm, $\mathrm{Nrd}(zw) = \mathrm{Nrd}(z)\mathrm{Nrd}(w)$,
*is* Euler's four-square identity. Working over a rational function field in eight
variables makes it an identity of polynomials rather than a numerical coincidence.

```python
P = PolynomialRing(QQ, ['x0','x1','x2','x3','y0','y1','y2','y3'])
F = P.fraction_field()
x0,x1,x2,x3,y0,y1,y2,y3 = F.gens()

Bg.<I,J,K> = QuaternionAlgebra(F, -1, -1)
zz = x0 + x1*I + x2*J + x3*K
ww = y0 + y1*I + y2*J + y3*K

print("Nrd(zw) == Nrd(z)Nrd(w) ?", bool((zz*ww).reduced_norm() == zz.reduced_norm()*ww.reduced_norm()))
print()
print("(x0^2+x1^2+x2^2+x3^2)(y0^2+y1^2+y2^2+y3^2) =")
show(P((zz*ww).reduced_norm().numerator()).factor())
print("\nand the four component squares on the left:")
for c in (zz*ww).coefficient_tuple():
    print("   ", P(c.numerator()))
```

## 2. General $(a,b)$ and where they ramify

For $a,b \in \mathbb{Q}^\times$, $\left(\frac{a,b}{\mathbb{Q}}\right)$ is a quaternion
algebra. By Albert–Brauer–Hasse–Noether it is determined up to isomorphism by its set of
ramified places, which is finite of even cardinality; the local invariant at $v$ is the
Hilbert symbol $(a,b)_v$. The **discriminant** is the product of the ramified *finite*
primes.

```python
rows = []
for (a, b) in [(-1,-1), (-1,1), (2,3), (-1,-7), (6,-35), (5,-3), (-2,-5), (13,17)]:
    A = QuaternionAlgebra(QQ, a, b)
    rows.append([f"({a},{b})", A.discriminant(), A.ramified_primes(),
                 "division" if A.is_division_algebra() else "M_2(Q)"])
table(rows, header_row=["(a,b)", "disc", "ramified primes", "type"])
```

Hilbert's reciprocity law: $\prod_v (a,b)_v = 1$ over all places, the real place
included. Sage writes the real place as `p = -1`.

```python
a, b = 6, -35
places = [-1] + prime_range(300)
syms = {p: hilbert_symbol(a, b, p) for p in places}
bad = [p for p, s in syms.items() if s == -1]

print("places where (a,b)_v = -1 :", bad, "   (-1 denotes the real place)")
print("product over all places   :", prod(syms.values()))
print("even number of them?      :", len(bad) % 2 == 0)
print("Sage's ramified_primes()  :", QuaternionAlgebra(a, b).ramified_primes())
```

Same ramification set $\Rightarrow$ isomorphic. Let us find a second symbol presenting
Hamilton's algebra rather than hardcoding one.

```python
H = QuaternionAlgebra(QQ, -1, -1)
found = []
for a in range(-8, 9):
    for b in range(-8, 9):
        if a == 0 or b == 0 or (a, b) == (-1, -1):
            continue
        A = QuaternionAlgebra(QQ, a, b)
        if A.ramified_primes() == H.ramified_primes() and A.is_division_algebra():
            if H.is_isomorphic(A):
                found.append((a, b))
print("symbols isomorphic to (-1,-1):", found[:10], "...")
print("check one:", QuaternionAlgebra(QQ, *found[0]).is_isomorphic(H))
print("a non-example:", H.is_isomorphic(QuaternionAlgebra(QQ, 2, 3)),
      "  (ramified at", QuaternionAlgebra(QQ,2,3).ramified_primes(), "vs", H.ramified_primes(), ")")
```

## 3. Splitting: zero divisors and reduction mod $p$

A quaternion algebra is either a division algebra or $M_2$ of the base field. In the
split case there are zero divisors, and they are easy to write down: any $z \neq 0$ with
$\mathrm{Nrd}(z) = 0$ is one, since $z\bar z = \mathrm{Nrd}(z) = 0$.

```python
S.<i2,j2,k2> = QuaternionAlgebra(QQ, 1, -1)
print("split?", S.is_matrix_ring(), " ramified at:", S.ramified_primes())

e = (1 + i2)/2
print("e =", e, "| e^2 == e ?", e*e == e, "| Nrd(e) =", e.reduced_norm())
print(f"zero divisor: ({e}) * ({1-e}) = {e*(1-e)}")
```

For a prime $p$ not dividing the discriminant, $B \otimes \mathbb{Q}_p \cong M_2(\mathbb{Q}_p)$.
Sage produces an explicit splitting map to $M_2(\mathbb{F}_p)$.

```python
B.<i,j,k> = QuaternionAlgebra(-1, -1)
for p in [3, 5, 7]:
    phi = B.modp_splitting_map(p)
    ok = all(phi(u*v) == phi(u)*phi(v) for u in [i, j, k, 1+i] for v in [i, j, k, 1-j])
    print(f"p = {p}: phi(i) and phi(j) over GF({p}), multiplicative: {ok}")
    print(phi(i), phi(j))
    print()
```

## 4. Orders, units, and class numbers

The **Lipschitz order** $\mathbb{Z}\langle i,j\rangle$ is not maximal; the **Hurwitz
order**, obtained by adjoining $\omega = \tfrac{1}{2}(1+i+j+k)$, is. Discriminants 4 and
2 respectively.

```python
O = B.maximal_order()
L = B.quaternion_order([1, i, j, k])
print("Hurwitz  basis:", O.basis(), " disc:", O.discriminant())
print("Lipschitz basis:", L.basis(), " disc:", L.discriminant())
print("Lipschitz contained in Hurwitz?", all(u in O for u in L.basis()))
```

```python
from itertools import product as iproduct
half = [QQ(c)/2 for c in [-2,-1,0,1,2]]
cands = [c0 + c1*i + c2*j + c3*k for (c0,c1,c2,c3) in iproduct(half, repeat=4)]
units_H = [u for u in cands if u.reduced_norm() == 1 and u in O]
units_L = [u for u in units_H if u in L]

print("Hurwitz units  :", len(units_H), " (the binary tetrahedral group 2T)")
print("Lipschitz units:", len(units_L), sorted(str(u) for u in units_L))
print("all of norm 1  :", {u.reduced_norm() for u in units_H})
```

Class numbers of maximal orders, via Brandt modules. For the algebra ramified exactly at
$\{p,\infty\}$ the class number is $\dim$ of the Brandt module $B(p,1)$.

```python
rows = [[p, QuaternionAlgebra(p).ramified_primes(), BrandtModule(p).dimension()]
        for p in prime_range(2, 60)]
table(rows, header_row=["p", "ramified", "class number"])
```

## 5. Quaternion algebras over number fields

Over a general number field the ramified set includes real places. Sage's
`ramified_places()` returns the pair (finite primes, real embeddings).

```python
R.<x> = QQ[]
K.<w> = NumberField(x^2 - 5)
A = QuaternionAlgebra(K, -1, -1)
fin, inf = A.ramified_places()
print(A)
print("finite ramified primes:", fin)
print("ramified real places  :", len(inf), "of", len(K.real_embeddings()))
print("division algebra?", A.is_division_algebra())

A2 = QuaternionAlgebra(K, -1, w)
fin2, inf2 = A2.ramified_places()
print()
print(A2)
print("finite ramified primes:", [(P.gens_reduced()[0], P.norm()) for P in fin2], "(generator, norm)")
print("ramified real places  :", len(inf2))
```

## 6. Cyclic algebras of arbitrary degree

Sage has no cyclic algebra class, so here is one.

Let $K/F$ be cyclic of degree $n$ with $\mathrm{Gal}(K/F) = \langle\sigma\rangle$, and
$b \in F^\times$. The cyclic algebra is
$$A = (K/F,\sigma,b) = \bigoplus_{r=0}^{n-1} K u^r, \qquad
  u x = \sigma(x)\,u \ \ (x \in K), \qquad u^n = b,$$
a central simple $F$-algebra of dimension $n^2$. For $n = 2$, $K = F(\sqrt a)$ recovers
the quaternion algebra $(a,b)$.

To compute we use the splitting $A \otimes_F K \cong M_n(K)$ explicitly:
$$x \longmapsto D_x := \mathrm{diag}\big(x, \sigma^{-1}(x), \dots, \sigma^{-(n-1)}(x)\big),
\qquad
u \longmapsto U := \begin{pmatrix} & & & b \\ 1 & & & \\ & \ddots & & \\ & & 1 & \end{pmatrix}.$$
One checks $U D_x = D_{\sigma(x)} U$ and $U^n = b\cdot 1$, so this really is a
representation. The **reduced norm and trace** are then just $\det$ and $\mathrm{tr}$ of
the resulting $n \times n$ matrix — and they land back in $F$.

> The twist by $\sigma^{-1}$ rather than $\sigma$ in $D_x$ matters for $n \geq 3$; for
> $n = 2$ it is invisible since $\sigma^{-1} = \sigma$.

```python
class CyclicAlgebra:
    r'''The cyclic algebra (K/F, sigma, b), F = QQ, K/QQ cyclic of degree n.

    Elements are lists [x_0, ..., x_{n-1}] of elements of K, meaning sum x_r u^r.
    '''
    def __init__(self, K, sigma, b):
        if not (K.is_galois() and K.galois_group().is_cyclic()):
            raise ValueError("K/QQ must be cyclic")
        self.K, self.sigma, self.b, self.n = K, sigma, K(b), K.degree()

    def sig(self, r, x):
        x = self.K(x)
        for _ in range(r % self.n):
            x = self.sigma(x)
        return x

    def mul(self, z, w):
        n, K = self.n, self.K
        out = [K(0)]*n
        for r in range(n):
            if z[r] == 0: continue
            for s in range(n):
                if w[s] == 0: continue
                t = z[r] * self.sig(r, w[s])
                if r + s >= n:
                    t *= self.b
                out[(r+s) % n] += t
        return out

    def U(self):
        M = matrix(self.K, self.n, self.n)
        for r in range(self.n - 1):
            M[r+1, r] = 1
        M[0, self.n-1] = self.b
        return M

    def D(self, x):
        return diagonal_matrix(self.K, [self.sig(-r, x) for r in range(self.n)])

    def rho(self, z):
        '''The embedding A -> M_n(K).'''
        U, Ur = self.U(), identity_matrix(self.K, self.n)
        M = matrix(self.K, self.n, self.n)
        for r in range(self.n):
            M += self.D(z[r]) * Ur
            Ur = Ur * U
        return M

    def nrd(self, z): return self.rho(z).det()
    def trd(self, z): return self.rho(z).trace()
    def is_unit(self, z): return self.nrd(z) != 0

print("defined CyclicAlgebra")
```

### Sanity check: degree 2 must reproduce Sage's quaternions

With $K = \mathbb{Q}(\sqrt a)$ and $\sigma$ the conjugation, the dictionary is
$i = \sqrt a$, $j = u$, $k = ij = \sqrt a\, u$. We compare reduced norms and traces
against `QuaternionAlgebra(a,b)` on random elements, and check that $\rho$ is really
multiplicative.

```python
R.<x> = QQ[]
set_random_seed(1)
for (a, b) in [(-1,-1), (2,3), (-1,-7), (5,-3)]:
    K.<w> = NumberField(x^2 - a)
    sigma = [s for s in K.automorphisms() if s(w) != w][0]
    A = CyclicAlgebra(K, sigma, b)
    Bq.<i,j,k> = QuaternionAlgebra(QQ, a, b)

    nrd_ok = trd_ok = True
    for _ in range(30):
        c = [QQ.random_element() for _ in range(4)]
        z = [K(c[0]) + c[1]*w, K(c[2]) + c[3]*w]
        q = c[0] + c[1]*i + c[2]*j + c[3]*k
        nrd_ok &= (A.nrd(z) == q.reduced_norm())
        trd_ok &= (A.trd(z) == q.reduced_trace())

    hom_ok = True
    for _ in range(10):
        z = [K.random_element(), K.random_element()]
        v = [K.random_element(), K.random_element()]
        hom_ok &= (A.rho(A.mul(z, v)) == A.rho(z)*A.rho(v))

    print(f"(a,b) = ({a:>3},{b:>3}):  Nrd agrees {nrd_ok},  Trd agrees {trd_ok},  rho is a homomorphism {hom_ok}")
```

### A degree 3 example

Take $K$ the cyclic cubic field of conductor 9, i.e. $\mathbb{Q}[x]/(x^3-3x-1)$
(this is $\mathbb{Q}(\zeta_9)^+$), and vary $b$.

For $n$ prime, $(K/F,\sigma,b)$ is **split iff $b$ is a norm from $K$**, and a division
algebra otherwise. When $b = N_{K/F}(\beta)$ the element $\beta - u$ is a nonzero
element of reduced norm zero: an explicit zero divisor witnessing the splitting.

```python
K.<a> = NumberField(x^3 - 3*x - 1)
print("Galois?", K.is_galois(), " cyclic?", K.galois_group().is_cyclic(),
      " disc:", K.discriminant(), " class number:", K.class_number())
sigma = [s for s in K.automorphisms() if s(a) != a][0]
print("sigma(a) =", sigma(a), "  sigma^3 = id ?", sigma(sigma(sigma(a))) == a)

rows = []
for b in [2, 3, 5, 7, 8, 11]:
    A = CyclicAlgebra(K, sigma, b)
    is_nm, beta = QQ(b).is_norm(K, element=True)
    if is_nm:
        witness = A.nrd([beta, K(-1), K(0)])          # Nrd(beta - u)
        note = f"beta = {beta}, Nrd(beta - u) = {witness}"
    else:
        note = "no norm: division algebra"
    rows.append([b, "split" if is_nm else "division", note])
table(rows, header_row=["b", "(K/Q,sigma,b)", "witness"])
```

```python
# The division case b = 2: reduced norms of nonzero elements never vanish.
A2 = CyclicAlgebra(K, sigma, 2)
set_random_seed(0)
norms = []
for _ in range(200):
    z = [K.random_element() for _ in range(3)]
    if all(c == 0 for c in z): continue
    norms.append(A2.nrd(z))
print("all reduced norms rational?", all(nm in QQ for nm in norms))
print("any zero (would contradict division)?", any(nm == 0 for nm in norms))
print("sample:", norms[:6])

z = [K(1) + a, K(2), K(a^2)]
print()
print("z          =", z, " (meaning x0 + x1 u + x2 u^2)")
print("Nrd(z)     =", A2.nrd(z), "   Trd(z) =", A2.trd(z))
print("rho(z) =");  show(A2.rho(z))
print("charpoly of rho(z) (the reduced characteristic polynomial):")
show(A2.rho(z).charpoly())
```

## 7. Coda: odd valuations and ramification

This ties back to the question about $v_p(f(n))$ being odd. For $a \in \mathbb{Q}^\times$
and $m \in \mathbb{Z}$, the Hilbert symbol $(a,m)_p$ at an odd prime $p \nmid a$ depends
on $m$ only through the parity of $v_p(m)$. So the primes at which the specialised
algebra $(a, f(n))$ can ramify are exactly those with $v_p(f(n))$ odd — and the theorem
says there are infinitely many such $p$ precisely when $f$ is not a constant times a
square.

Sage's `squarefree_decomposition` exhibits the relevant invariant directly: the
odd-multiplicity part of $f$.

```python
R.<x> = QQ[]
for f in [x^2, x^2*(x+1), 12*(x^2+1)^2, x^3, (x^2+1)^2*(x-1)^3]:
    dec = f.squarefree_decomposition()
    odd = prod([g^e for g, e in dec if e % 2 == 1], R(1))
    print(f"f = {str(factor(f)):<28}  odd-multiplicity part = {factor(odd) if odd != 1 else 1}")
```

```python
def odd_valuation_primes(f, N=80):
    S = set()
    for n in range(1, N):
        m = ZZ(f(n))
        if m == 0: continue
        S |= {p for p, e in factor(m) if e % 2 == 1}
    return S

def ramified_of_specialisations(a, f, N=80):
    S = set()
    for n in range(1, N):
        m = ZZ(f(n))
        if m == 0: continue
        S |= set(QuaternionAlgebra(QQ, a, m).ramified_primes())
    return S

def fmt(S, k=12):
    return f"{sorted(S)[:k]}{' ...' if len(S) > k else ''}   (count {len(S)})"

for f in [x^2, 12*(x^2+1)^2, x^2*(x+1)]:
    S = odd_valuation_primes(f)
    T = ramified_of_specialisations(-1, f)
    print(f"f = {factor(f)}")
    print(f"   v_p(f(n)) odd for some n <= 80       : {fmt(S)}")
    print(f"   ramified primes of (-1, f(n))        : {fmt(T)}")
    print(f"   away from 2, ramified subset of odd? : {set(p for p in T if p != 2) <= S}")
    print(f"   is 2 ramified for some n?            : {2 in T}   (v_2 odd somewhere? {2 in S})")
    print()
```

Reading the output: $x^2$ gives nothing at all (the parity of $v_p$ is constant in $n$),
and $12(x^2+1)^2$ gives exactly $\{3\}$ — the primes dividing the content $12 = 2^2\cdot 3$
to an odd power. Only $x^2(x+1)$, which has an irreducible factor of odd multiplicity,
produces a set that grows without bound.

Away from $2$ the ramified primes of $(-1, f(n))$ are exactly the odd-valuation primes
that are additionally $\equiv 3 \pmod 4$, which is where the second Hilbert symbol
condition bites. The prime $2$ is genuinely exceptional and the containment fails there:
$(-1, m)_2$ depends on $m$ modulo $8$, not merely on the parity of $v_2(m)$. That is why
$2$ shows up as ramified for $12(x^2+1)^2$ even though $v_2(f(n))$ is even for every $n$.
